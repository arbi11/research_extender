# Topology Optimization Through Sequential Decision-Making

## Introduction

Topology optimization (TO) represents an advanced computational design methodology that creates efficient structural configurations through systematic material distribution, satisfying performance requirements while respecting constraints. Over the past three decades, TO has transformed design practices across aerospace {cite}`zhu2016topology`, automotive {cite}`jankovics2019customization`, civil engineering {cite}`jewett2019topology`, and mechanical engineering {cite}`guanghui2020aerospace,ZHU202191`. Unlike traditional design approaches that rely primarily on engineer intuition and incremental modifications to existing configurations, TO algorithms generate innovative, high-performance solutions through principled mathematical optimization rooted in physics-based simulation.

The transformative potential of TO manifests dramatically in real-world applications. General Electric's optimization of an aircraft engine mounting bracket, shown in Figure {numref}`fig-ge-bracket`, achieved an 84% weight reduction compared to the conventional design {cite}`morgan2014ge`. This seemingly modest modification to a single component translated to approximately \$31 million in annual savings for airlines through improved fuel efficiency. Similarly, Airbus employed TO to design the droop nose ribs for the A380 aircraft, achieving required mechanical performance while substantially reducing structural weight {cite}`krog2002application`. These examples illustrate how TO enables engineers to discover designs that would be extraordinarily difficult - if not impossible - to conceive through traditional parametric approaches.

The significance of TO has intensified dramatically with the development and maturation of Additive Manufacturing (AM) technologies. Manufacturing constraints that historically limited TO applications - particularly the difficulty of fabricating complex internal geometries through conventional subtractive processes - have been substantially relaxed by AM capabilities. Industries increasingly employ advanced design methods including topology optimization and generative design {cite}`atkinson2019manufacturing,Coop2018` to develop next-generation products. This convergence of computational design optimization and advanced manufacturing creates unprecedented opportunities for innovation in electrical machine design.

## The Imperative for Electrical Machine Optimization

Electrical machines constitute the foundation of modern industrial infrastructure and power systems. These devices serve as the core components of industrial machinery, appliances, and electromechanical systems {cite}`lei2017review`. The global electricity consumption attributed to electric motors reaches 43-46% of total generation worldwide {cite}`waide2011energy`, as illustrated in Figure {numref}`fig-electricity-consumption`. This massive energy utilization translates to approximately 6,040 megatonnes of CO₂ emissions annually, establishing electric motor efficiency as the single largest opportunity for electricity conservation and environmental impact reduction among all end-use categories.

The imperative for high-efficiency motor design intensifies with accelerating global electrification trends. As discussed in Chapter 1, transportation electrification—progressing from 17,000 electric vehicles globally in 2010 to over 14 million units sold in 2023—demands tens of millions of high-performance traction motors annually. International Energy Agency projections indicate that high-efficiency motors will dominate global electrical machine markets through 2030, driven by regulatory mandates (IE4/IE5 standards), economic competitiveness (energy costs representing 20-40% of industrial facility operating expenses), and climate commitments (Paris Agreement compliance requiring 30-40% of emission reductions from transportation and industrial electrification) {cite}`IEA_WEO_2019,unfccc2015paris`.

In the context of global sustainability imperatives, the demand for environmentally conscious design alternatives continues to accelerate. Future electrical machines must satisfy not only traditional performance metrics—electromagnetic efficiency, torque density, power factor—but also environmental impact criteria across their entire lifecycle. This expanding scope of design optimization necessarily increases problem complexity as additional engineering domains and constraints enter consideration: electromagnetics, structural mechanics, thermal management, acoustic emissions, and manufacturing feasibility {cite}`lei2017review`. Consequently, design and manufacturing organizations increasingly employ TO in pursuit of designs that simultaneously satisfy performance, sustainability, and economic objectives.

## The Challenge of Design Space Exploration

The theoretical foundations of topology optimization trace to Michell's 1904 work on optimal truss layouts {cite}`lviiiagm1904limits`, with renewed interest emerging in the 1950s-1960s focused on structural applications {cite}`cox1958structures,hemp1958theory`. Significant publications in layout theory proliferated through the 1960s and 1970s {cite}`prager1968problems,dobbs1969optimization`. The first application of TO to electromagnetic field problems appeared in 1996, when Dyck and colleagues proposed the Optimized Material Distribution method {cite}`dyck1996automated`.

Despite this substantial research history and proven industrial value, TO faces fundamental computational challenges that constrain its application to complex electromagnetic device design. The transformation of design optimization into a material distribution problem—while offering unprecedented design freedom—creates combinatorial search spaces of extraordinary size. For a design domain discretized into *n* elements, each capable of assuming one of *k* material states, the complete search space contains *k^n* possible configurations. A modest 30×30 grid with binary material assignment (air or ferromagnetic core) yields 2^900 ≈ 10^271 possible topologies—a number that dwarfs the estimated 10^80 atoms in the observable universe.

Traditional optimization approaches struggle with this combinatorial explosion. Deterministic gradient-based methods, while computationally efficient per iteration, frequently converge to local optima in the highly non-convex electromagnetic TO landscape. Stochastic methods—particularly Evolutionary Algorithms (EA) such as Genetic Algorithms—provide superior global search capability and robustness against local minima {cite}`li2019numerical`. However, EA computational demands scale poorly with problem dimensionality. A typical EA-based TO run for an electromagnetic device requires 10,000-40,000 finite element analysis evaluations. At 2-8 hours per 2D quasi-static FEA and days per 3D transient analysis, comprehensive TO exploration can demand weeks to months of continuous computation {cite}`lei2017review`—timelines incompatible with industrial product development cycles.

## Sizing, Shape, and Topology Optimization

Design optimization methods span a spectrum of geometric freedom (Figure {numref}`fig-to-comparison`). *Sizing* optimization adjusts scalar parameters within a fixed template. *Shape* optimization modifies boundaries while preserving topological connectivity. *Topology* optimization removes all such constraints — the number and layout of holes, members, and material regions vary freely. This freedom is precisely what makes TO powerful and computationally expensive. Section 2 provides a full taxonomy of existing methods and how each handles the resulting combinatorial challenge.

## Sequential Decision-Making: The MDP Framework

The computational challenges inherent in discrete topology optimization—particularly the combinatorial explosion of possible material distributions and the persistent checkerboard artifacts in density-based and ON/OFF methods—motivate the investigation of fundamentally different problem formulations. This chapter introduces a novel approach: reformulating TO as a sequential decision-making problem amenable to Reinforcement Learning (RL) algorithms.

Traditional TO methods treat material distribution as a simultaneous optimization problem: determine the optimal value of all design variables (material presence/absence for each grid cell) at once. This formulation, while conceptually straightforward, creates the massive search spaces that plague evolutionary algorithms. An alternative perspective emerges from recognizing that material distribution can be constructed sequentially: make a series of local placement decisions that collectively produce a complete topology.

This insight enables formulation as a **Markov Decision Process (MDP)**, the mathematical framework underlying reinforcement learning {cite}`sutton2018reinforcement`. In the MDP formulation:

- **States (S)**: Current binary occupancy of the design domain — the material distribution across all discretized cells at the current construction step
- **Actions (A)**: Local material placement decisions (controller movements in design space)
- **Transition function (P)**: Deterministic evolution of topology based on action selection
- **Reward function (R)**: Performance feedback based on electromagnetic simulation
- **Policy (π)**: Strategy for selecting actions that maximises cumulative reward

This sequential formulation provides several advantages. First, it naturally ensures *connectivity*: by constructing topologies through a moving controller that leaves a material trail, the resulting configurations are inherently connected without requiring filtering or penalty terms. Second, it enables application of powerful RL algorithms—particularly Q-learning and Deep Q-Networks—that have demonstrated remarkable success in complex sequential decision problems ranging from game playing to robotics {cite}`mnih2015human,silver2016mastering`. Third, the MDP structure provides a principled framework for balancing exploration (trying novel design strategies) against exploitation (refining known promising approaches), addressing a key challenge in stochastic optimization.

Importantly, this MDP formulation does not abandon the strengths of evolutionary algorithms. Genetic Algorithms can operate effectively within the sequential framework, evolving sequences of actions rather than complete material distributions. This enables direct comparison: does the sequential decision structure combined with RL provide advantages over evolutionary approaches operating in the same sequential space?

## Research Contributions

This chapter presents several interconnected contributions addressing the challenges of topology optimization for electromagnetic devices:

**1. Sequential Topology Optimization Framework (SeqTO-v1)**

We introduce a sequence-based TO methodology that constructs topologies through controller movements in discretized design space. The controller—characterised by size and associated material—moves according to action sequences (up, down, left, right) while depositing material along its path. This construction process inherently ensures connectivity, eliminating checkerboard patterns, isolated material islands, and thin connections that plague traditional density-based and ON/OFF methods. The framework requires neither filtering techniques nor penalty terms in the objective function to achieve manufacturable designs.

**2. MDP Formulation for Electromagnetic TO**

We formulate electromagnetic topology optimization as a Markov Decision Process, providing the mathematical foundation for applying reinforcement learning algorithms. The formulation specifies state representation (current topology configuration), action space (controller movements), reward structure (electromagnetic performance metrics from FEMM-based FEA), and termination conditions. This contribution establishes the first MDP framework specifically designed for electromagnetic device optimization.

**3. Application to C-Core Actuators and Synchronous Reluctance Motors**

We demonstrate the framework's efficacy through two benchmark electromagnetic design problems: core material distribution for C-core electromagnetic actuators and rotor flux barrier optimization for Synchronous Reluctance Motors (SynRM). The C-core actuator (288-cell design space, 24×12 half-symmetric) serves as the primary experimental benchmark, achieving 70 N force (75% improvement over the 40 N conventional baseline). The SynRM rotor (25-cell design space, 5×5 quarter geometry) demonstrates SeqTO-v1 generalises across device classes, achieving 3.40 Nm average torque (+36% over the 2.5 Nm conventional baseline).

**4. Q-Learning Implementation and Performance Analysis**

We implement Q-learning algorithms operating within the MDP framework, demonstrating that reinforcement learning can discover high-performing topologies through interaction with the FEMM electromagnetic simulation environment. Performance analysis quantifies convergence characteristics, computational requirements (15–20% fewer FE evaluations than GA), and solution quality relative to evolutionary baselines.

**5. Comparative Evaluation: RL versus Genetic Algorithms**

We conduct systematic comparisons between reinforcement learning and genetic algorithm approaches, both operating within the SeqTO-v1 sequential framework. This controlled comparison isolates the impact of the learning strategy from the sequential formulation, providing insights into the relative strengths of RL-based and EA-based approaches for electromagnetic TO.

**6. Extension Framework for Deep Reinforcement Learning**

The chapter concludes by establishing the foundation for deep RL extensions (addressed in Chapter 5). The MDP formulation and sequential construction process enable application of function approximation methods—particularly Deep Q-Networks—that can generalise across similar design problems, potentially reducing computational burden when optimising related electromagnetic device families.

## Chapter Organisation

This chapter is organised into seven sections:

**Section 1** (this section) motivates the research through real-world impact examples and introduces the SeqTO-v1 sequential decision-making framework as the core contribution.

**Section 2** surveys topology optimisation methodologies — homogenisation, density-based SIMP, level-set, and discrete ON/OFF methods — and positions the sequential MDP approach within the research landscape.

**Section 3** develops the MDP mathematical framework and Q-learning algorithm together: Bellman equations, value iteration, the SeqTO-v1 controller-based construction process, and the complete Q-learning implementation for electromagnetic TO.

**Section 4** covers both the C-core electromagnetic actuator and the Synchronous Reluctance Motor (SynRM) rotor environments under the SeqTO-v1 formulation: design domain geometry, FEMM-based FEA setup, MDP state/action/reward specification, and material budget constraints for each device.

**Section 5** presents the genetic algorithm as a comparison baseline within the SeqTO-v1 sequential framework, specifying action-sequence chromosome encoding, crossover/mutation operators, and the MATLAB Global Optimization Toolbox parameters used in experiments.

**Section 6** presents the MDP implementation framework: the production `QLearningAgent` and `QLearningTrainer` from `Chp4_MDP_TO/src/q_learning/`, and the `CcoreFemmEnv` interface from `Chp5_DRL_TO/src/femm_environment/`. All code shown integrates directly with the FEMM magnetostatic solver for physics-based reward evaluation.

**Section 7** presents experimental results (C-core: 70 N / +75%; SynRM: 3.40 Nm / +36%), FE efficiency analysis (4–5× fewer evaluations than ON/OFF GA), and future research directions including deep RL, transfer learning, and multi-physics integration.

Through this progression the chapter demonstrates that reformulating topology optimisation as sequential decision-making enables reinforcement learning algorithms that discover high-performing, manufacturable designs efficiently—with SeqTO-v1 achieving both superior performance and substantial computational savings over conventional approaches.
