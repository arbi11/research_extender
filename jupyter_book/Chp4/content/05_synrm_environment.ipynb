# 5. Synchronous Reluctance Motor Environment

## 5.1 Introduction to SynRM Design

Synchronous Reluctance Motors (SynRMs) represent a compelling class of electrical machines that have gained significant industrial attention due to their simple structure, low cost, and high efficiency. Unlike permanent magnet motors, SynRMs generate torque solely through reluctance variation, making them attractive for applications where cost and reliability are paramount.

### 5.1.1 Working Principle

SynRMs operate on the fundamental principle of **reluctance torque**: the rotor tends to align itself in a position of minimum magnetic reluctance with respect to the stator's rotating magnetic field. This phenomenon can be expressed mathematically as:

$$T = \frac{3}{2} \cdot \frac{P}{2} \cdot \left(\frac{V_s}{\omega_s}\right)^2 \cdot \left(\frac{1}{L_q} - \frac{1}{L_d}\right) \cdot \sin(2\delta)$$

Where:
- $T$: Electromagnetic torque
- $P$: Number of poles
- $V_s$: Stator voltage
- $\omega_s$: Synchronous angular velocity
- $L_d, L_q$: Direct and quadrature axis inductances
- $\delta$: Torque angle

### 5.1.2 Advantages over Conventional Motors

SynRMs offer several compelling advantages:

**Efficiency Benefits**:
- **No rotor losses**: Absence of rotor windings eliminates copper losses
- **High efficiency**: Recent designs achieve up to 99.05% efficiency
- **Lower operating temperature**: Reduced thermal stress on components

**Manufacturing Advantages**:
- **Simple construction**: No permanent magnets, brushes, or slip rings
- **Low cost**: Elimination of expensive rare-earth materials
- **Robust design**: Suitable for harsh industrial environments

**Performance Characteristics**:
- **High power density**: 2-4 times higher than induction motors
- **Compact footprint**: 1-2 times smaller frame size
- **Maintenance-free**: Long operational life with minimal servicing

### 5.1.3 Industrial Applications

SynRMs find applications in diverse industrial sectors:
- **Appliances**: Washing machines, HVAC systems
- **Automotive**: Electric vehicle auxiliary systems
- **Industrial**: Pumps, fans, compressors
- **Precision devices**: Analog meters, hard disk drives

## 5.2 Rotor Design Fundamentals

### 5.2.1 Flux Barrier Concept

The key to SynRM performance lies in the strategic placement of **flux barriers** in the rotor structure. These barriers guide magnetic flux along preferred paths, creating the necessary reluctance variation for torque generation.

**Flux Barrier Functions**:
- **Direct flux guidance**: Channel flux through high-permeability iron paths
- **Quadrature flux blocking**: Prevent flux leakage between poles
- **Saliency creation**: Maximize $L_d/L_q$ ratio for torque production
- **Structural integrity**: Maintain mechanical strength at high speeds

### 5.2.2 Rotor Types

Three main SynRM rotor configurations exist:

**1. Simple Salient Pole Rotor**:
- **Design**: Projecting poles with air gaps between
- **Advantages**: Simple manufacturing, robust construction
- **Limitations**: Lower saliency ratio, higher torque ripple

**2. Transverse Laminated Rotor**:
- **Design**: Stacked steel sheets with flux barriers
- **Advantages**: Better flux control, reduced eddy currents
- **Limitations**: Complex manufacturing, alignment requirements

**3. Axially Laminated Rotor**:
- **Design**: Axial flux barriers with layered construction
- **Advantages**: High saliency, excellent performance
- **Limitations**: Very complex manufacturing, high cost

### 5.2.3 Design Parameters

Key geometric parameters affecting SynRM performance:

**Barrier Geometry**:
- **Number of barriers**: Typically 2-4 barriers per pole
- **Barrier width**: Trade-off between flux blocking and structural integrity
- **Barrier shape**: Linear, curved, or complex geometries
- **Barrier position**: Radial location and angular extent

**Rotor Dimensions**:
- **Outer diameter**: Determines torque capacity
- **Inner diameter**: Affects mechanical robustness
- **Stack length**: Influences power rating
- **Air gap**: Critical for performance and manufacturability

## 5.3 SynRM for Topology Optimization

### 5.3.1 Optimization Objectives

The primary objective in SynRM topology optimization is **torque maximization**:

$$\max_{\mathbf{p}} T_{\text{avg}}(\mathbf{p})$$

Subject to constraints:
$$K_1 \leq \sum_{j=1}^{n} (p_j = 0) \leq K_2$$

Where:
- $\mathbf{p}$: Material distribution vector ($p_j = 1$ for iron, $p_j = 0$ for air)
- $T_{\text{avg}}$: Average electromagnetic torque
- $K_1, K_2$: Minimum and maximum air (flux barrier) constraints

### 5.3.2 Design Space Discretization

For topology optimization, the rotor design space is discretized into a grid:

**Grid Resolution Considerations**:
- **Coarse grids (5×5)**: Fast computation, limited design freedom
- **Medium grids (10×10)**: Balanced approach, reasonable detail
- **Fine grids (20×20+)**: High detail, computationally expensive

**Symmetry Exploitation**:
For a 4-pole, 3-phase design, only **one-quarter** of the rotor needs to be optimized:
$$\text{Full rotor} = 4 \times \text{Quarter design}$$

This reduces computational complexity by a factor of 4 while maintaining design completeness.

### 5.3.3 Material Representation

**Binary Material Model**:
$$\rho_{i,j} = \begin{cases} 1 & \text{M-19 steel (magnetic material)} \\ 0 & \text{Air (flux barrier)} \end{cases}$$

**Properties**:
- **M-19 Steel**: $\mu_r \approx 2000$, high saturation flux density
- **Air**: $\mu_r = 1$, non-magnetic barrier

## 5.4 SynRM Environment for Reinforcement Learning

### 5.4.1 State Space Design

The state represents the current rotor topology:

**Topological State**:
$$s_t = \begin{bmatrix} \rho_{0,0} & \rho_{0,1} & \cdots & \rho_{0,n-1} \\ \rho_{1,0} & \rho_{1,1} & \cdots & \rho_{1,n-1} \\ \vdots & \vdots & \ddots & \vdots \\ \rho_{n-1,0} & \rho_{n-1,1} & \cdots & \rho_{n-1,n-1} \end{bmatrix}$$

**Performance State** (augmented):
- **Current torque**: $T_{\text{current}}$
- **Torque improvement**: $\Delta T = T_{\text{current}} - T_{\text{baseline}}$
- **Material usage**: $M_\text{ratio} = \frac{\sum \rho_{i,j}}{n^2}$

### 5.4.2 Action Space Design

Actions represent design modifications to the rotor topology:

**Primary Actions**:
1. **Add Flux Barrier**: $a_{\text{add}}(i,j)$: Change $\rho_{i,j}$ from 1 to 0
2. **Remove Flux Barrier**: $a_{\text{remove}}(i,j)$: Change $\rho_{i,j}$ from 0 to 1
3. **Toggle Cell**: $a_{\text{toggle}}(i,j)$: Flip material state at position $(i,j)$

**Secondary Actions** (advanced):
- **Add Barrier Line**: Create linear flux barrier
- **Add Barrier Curve**: Create curved flux barrier
- **Modify Barrier Width**: Change existing barrier dimensions

### 5.4.3 Reward Function Design

The reward function guides the optimization toward high-performance designs:

**Primary Reward Component**:
$$r_{\text{torque}} = w_T \cdot \frac{T_{\text{new}} - T_{\text{old}}}{T_{\text{baseline}}}$$

**Constraint Penalty Components**:
- **Material Usage Penalty**: $r_{\text{material}} = -w_m \cdot |M_\text{ratio} - M_\text{target}|$
- **Connectivity Penalty**: $r_{\text{connect}} = -w_c \cdot C_\text{violations}$
- **Manufacturing Penalty**: $r_{\text{manufacturing}} = -w_{\text{manu}} \cdot M_\text{violations}$

**Total Reward**:
$$r = r_{\text{torque}} + r_{\text{material}} + r_{\text{connect}} + r_{\text{manufacturing}}$$

## 5.5 Physical Simulation Integration

### 5.5.1 Electromagnetic Analysis

Accurate torque calculation requires electromagnetic field analysis:

**Finite Element Method (FEM)**:
- **Governing equation**: $\nabla \times (\frac{1}{\mu} \nabla \times \mathbf{A}) = \mathbf{J}$
- **Boundary conditions**: Appropriate stator and rotor conditions
- **Mesh refinement**: Fine mesh in air gap and flux barriers

**Magnetic Circuit Analysis** (simplified):
- **Reluctance network**: Analogous to electric circuits
- **Flux paths**: Main and leakage flux paths
- **Inductance calculation**: $L_d, L_q$ from circuit analysis

### 5.5.2 Performance Metrics

**Primary Performance Indicators**:
- **Average Torque**: $T_{\text{avg}} = \frac{1}{T} \int_0^T T(t) dt$
- **Torque Ripple**: $T_{\text{ripple}} = \frac{T_{\text{max}} - T_{\text{min}}}{T_{\text{avg}}}$
- **Efficiency**: $\eta = \frac{P_{\text{mechanical}}}{P_{\text{electrical}}}$
- **Power Factor**: $\cos \phi = \frac{P_{\text{real}}}{P_{\text{apparent}}}$

**Secondary Metrics**:
- **Saliency Ratio**: $\xi = \frac{L_d}{L_q}$
- **Core Losses**: $P_{\text{core}} = f(B, f, \text{material})$
- **Mechanical Stress**: Structural integrity validation

## 5.6 Design Constraints and Validation

### 5.6.1 Manufacturing Constraints

**Minimum Feature Size**:
- **Minimum barrier width**: $w_{\text{min}} \geq 1$ mm (manufacturing limit)
- **Minimum iron bridge**: $b_{\text{min}} \geq 0.5$ mm (structural requirement)
- **Corner radius**: $r_{\text{min}} \geq 0.5$ mm (stress concentration)

**Connectivity Requirements**:
- **Iron connectivity**: All iron regions must be connected to rotor core
- **Flux barrier continuity**: Barriers should not create isolated islands
- **Symmetry maintenance**: 4-fold symmetry must be preserved

### 5.6.2 Electromagnetic Constraints

**Flux Density Limits**:
- **Iron saturation**: $B_{\text{max}} \leq 1.8$ T for M-19 steel
- **Air gap flux**: $B_{\text{airgap}} \leq 1.2$ T (practical limit)

**Thermal Constraints**:
- **Temperature rise**: $\Delta T \leq 80$°C (insulation limit)
- **Hot spot prevention**: Uniform flux distribution

### 5.6.3 Validation Criteria

**Performance Validation**:
- **Torque requirement**: $T_{\text{design}} \geq T_{\text{specification}}$
- **Efficiency target**: $\eta \geq \eta_{\text{target}}$
- **Torque ripple limit**: $T_{\text{ripple}} \leq 10\%$ (for smooth operation)

**Manufacturing Validation**:
- **Structural analysis**: Stress analysis at rated speed
- **Vibration analysis**: Critical speed evaluation
- **Assembly verification**: Manufacturing feasibility check

```python
class SynRMOptimizationEnvironment:
    """SynRM topology optimization environment for RL"""
    
    def __init__(self, grid_size, parameters):
        self.grid_size = grid_size
        self.parameters = parameters
        self.reset_environment()
    
    def reset(self):
        """Reset environment to initial state"""
        self.topology = np.ones((self.grid_size, self.grid_size))
        self.current_performance = self.evaluate_design(self.topology)
        return self.get_state()
    
    def step(self, action):
        """Execute action and return observation"""
        old_topology = self.topology.copy()
        old_performance = self.current_performance
        
        # Apply action
        self.topology = self.apply_action(old_topology, action)
        
        # Evaluate new design
        new_performance = self.evaluate_design(self.topology)
        
        # Calculate reward
        reward = self.calculate_reward(old_performance, new_performance)
        
        # Check termination
        done = self.check_termination()
        
        return self.get_state(), reward, done, {
            'topology': self.topology.copy(),
            'performance': new_performance
        }
```

```````python
# SynRM environment visualization and demonstration
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Wedge, Rectangle
from IPython.display import display, Markdown
import matplotlib.patches as mpatches

class SynRMEnvironment:
    """Simplified SynRM environment for demonstration"""
    
    def __init__(self, grid_size=6, rotor_outer_dia=100, rotor_inner_dia=50):
        self.grid_size = grid_size
        self.rotor_outer_dia = rotor_outer_dia
        self.rotor_inner_dia = rotor_inner_dia
        self.reset()
    
    def reset(self):
        """Initialize with solid rotor (all iron)"""
        self.topology = np.ones((self.grid_size, self.grid_size))
        self.performance_history = []
        return self.get_state()
    
    def get_state(self):
        """Get current state representation"""
        return {
            'topology': self.topology.copy(),
            'performance': self.evaluate_performance(self.topology)
        }
    
    def step(self, action):
        """Execute action: toggle material at grid position"""
        i, j = action // self.grid_size, action % self.grid_size
        
        # Toggle material at position
        old_topology = self.topology.copy()
        self.topology[i, j] = 1 - self.topology[i, j]
        
        # Calculate performance change
        old_performance = self.evaluate_performance(old_topology)
        new_performance = self.evaluate_performance(self.topology)
        
        # Calculate reward
        reward = self.calculate_reward(old_performance, new_performance)
        
        self.performance_history.append(new_performance['torque'])
        
        return self.get_state(), reward, False, {
            'action': (i, j),
            'torque_change': new_performance['torque'] - old_performance['torque']
        }
    
    def evaluate_performance(self, topology):
        """Simplified performance evaluation"""
        # Simplified torque calculation based on flux barrier patterns
        torque = self.calculate_torque(topology)
        efficiency = self.calculate_efficiency(topology)
        torque_ripple = self.calculate_torque_ripple(topology)
        
        return {
            'torque': torque,
            'efficiency': efficiency,
            'torque_ripple': torque_ripple
        }
    
    def calculate_torque(self, topology):
        """Simplified torque calculation"""
        # Base torque for solid rotor
        base_torque = 2.0
        
        # Analyze flux barrier patterns
        # Check for radial flux barriers (good for torque)
        radial_bonuses = 0
        for j in range(self.grid_size):
            column = topology[:, j]
            # Look for vertical barriers (radial direction)
            if self.is_radial_barrier(column):
                radial_bonuses += 0.5
        
        # Check for proper iron bridges (structural integrity)
        bridge_bonus = self.count_iron_bridges(topology) * 0.2
        
        # Penalty for too much air (weak rotor)
        air_ratio = 1 - np.mean(topology)
        air_penalty = -1.0 if air_ratio > 0.6 else 0
        
        return base_torque + radial_bonuses + bridge_bonus + air_penalty
    
    def is_radial_barrier(self, column):
        """Check if column represents a radial flux barrier"""
        # Look for pattern: iron-air-iron (good flux barrier)
        transitions = np.sum(np.abs(np.diff(column)))
        return transitions == 2 and np.sum(column) >= 2  # At least 2 iron pieces
    
    def count_iron_bridges(self, topology):
        """Count iron bridges for structural integrity"""
        bridges = 0
        # Check outer ring
        bridges += np.sum(topology[0, :]) + np.sum(topology[-1, :])
        bridges += np.sum(topology[:, 0]) + np.sum(topology[:, -1])
        return bridges
    
    def calculate_efficiency(self, topology):
        """Simplified efficiency calculation"""
        # Efficiency based on flux path optimization
        base_efficiency = 0.92
        air_ratio = 1 - np.mean(topology)
        
        # Optimal air ratio for efficiency
        optimal_air_ratio = 0.3
        efficiency_loss = abs(air_ratio - optimal_air_ratio) * 0.1
        
        return base_efficiency - efficiency_loss
    
    def calculate_torque_ripple(self, topology):
        """Simplified torque ripple calculation"""
        # Ripple based on flux distribution uniformity
        uniformity = np.std(topology)
        return uniformity * 0.2  # Simplified relationship
    
    def calculate_reward(self, old_perf, new_perf):
        """Calculate reward based on performance improvement"""
        torque_improvement = new_perf['torque'] - old_perf['torque']
        efficiency_improvement = new_perf['efficiency'] - old_perf['efficiency']
        ripple_change = old_perf['torque_ripple'] - new_perf['torque_ripple']
        
        # Weighted reward
        reward = (
            2.0 * torque_improvement +
            5.0 * efficiency_improvement +
            1.0 * ripple_change
        )
        
        return reward
    
    def visualize_rotor(self, ax, topology=None, title="Rotor Topology"):
        """Visualize rotor topology"""
        if topology is None:
            topology = self.topology
        
        # Draw rotor outline
        outer_circle = Circle((0, 0), self.rotor_outer_dia/2, fill=False, edgecolor='black', linewidth=2)
        inner_circle = Circle((0, 0), self.rotor_inner_dia/2, fill=False, edgecolor='black', linewidth=2)
        ax.add_patch(outer_circle)
        ax.add_patch(inner_circle)
        
        # Draw topology grid
        grid_extent = (self.rotor_outer_dia - self.rotor_inner_dia) / 2
        cell_size = grid_extent / self.grid_size
        
        for i in range(self.grid_size):
            for j in range(self.grid_size):
                # Calculate cell position
                r = self.rotor_inner_dia/2 + (i + 0.5) * cell_size
                theta = (j + 0.5) * (np.pi / 2) / self.grid_size  # Quarter circle
                
                x = r * np.cos(theta)
                y = r * np.sin(theta)
                
                # Draw cell
                color = 'gray' if topology[i, j] == 1 else 'white'
                rect = Rectangle((x - cell_size/3, y - cell_size/3), 
                              cell_size/1.5, cell_size/1.5, 
                              facecolor=color, edgecolor='black', linewidth=0.5)
                ax.add_patch(rect)
        
        ax.set_xlim(-self.rotor_outer_dia/2 - 10, self.rotor_outer_dia/2 + 10)
        ax.set_ylim(-self.rotor_outer_dia/2 - 10, self.rotor_outer_dia/2 + 10)
        ax.set_aspect('equal')
        ax.set_title(title)
        ax.grid(True, alpha=0.3)

# Demonstrate SynRM environment
env = SynRMEnvironment(grid_size=5)

# Create figure with multiple visualizations
fig = plt.figure(figsize=(16, 12))

# 1. Initial rotor design
ax1 = plt.subplot(2, 3, 1, projection='polar')
env.visualize_rotor_polar(ax1, title="Initial Solid Rotor")

# 2. Simulated optimization episode
state = env.reset()
topology_steps = [state['topology'].copy()]
performance_steps = [state['performance']]
rewards = []

# Run optimization episode
actions = [7, 12, 18, 22, 13, 8, 17, 11]  # Predefined actions for demonstration

for step, action in enumerate(actions):
    state, reward, done, info = env.step(action)
    topology_steps.append(state['topology'].copy())
    performance_steps.append(state['performance'])
    rewards.append(reward)

# 3. Final optimized design
ax2 = plt.subplot(2, 3, 2, projection='polar')
env.visualize_rotor_polar(ax2, topology=topology_steps[-1], title="Optimized Rotor Design")

# 4. Performance progression
ax3 = plt.subplot(2, 3, 3)
torques = [p['torque'] for p in performance_steps]
efficiencies = [p['efficiency'] for p in performance_steps]
ax3_twin = ax3.twinx()

ax3.plot(range(len(torques)), torques, 'b-o', label='Torque (Nm)')
ax3_twin.plot(range(len(efficiencies)), efficiencies, 'r-s', label='Efficiency')

ax3.set_xlabel('Optimization Step')
ax3.set_ylabel('Torque (Nm)', color='b')
ax3_twin.set_ylabel('Efficiency', color='r')
ax3.tick_params(axis='y', labelcolor='b')
ax3_twin.tick_params(axis='y', labelcolor='r')
ax3.set_title('Performance Evolution')
ax3.grid(True, alpha=0.3)

# 5. Reward progression
ax4 = plt.subplot(2, 3, 4)
cumulative_rewards = np.cumsum(rewards)
ax4.bar(range(len(rewards)), rewards, alpha=0.7, color='green', label='Immediate Reward')
ax4.plot(range(len(cumulative_rewards)), cumulative_rewards, 'r-', linewidth=2, label='Cumulative Reward')
ax4.set_xlabel('Action')
ax4.set_ylabel('Reward')
ax4.set_title('Reward Progression')
ax4.legend()
ax4.grid(True, alpha=0.3)

# 6. Topology evolution
ax5 = plt.subplot(2, 3, 5)
final_topology = topology_steps[-1]
im = ax5.imshow(final_topology, cmap='binary', interpolation='nearest')
ax5.set_title('Final Topology (Grid View)')
ax5.set_xlabel('Angular Position')
ax5.set_ylabel('Radial Position')
plt.colorbar(im, ax=ax5, label='Material (1=Iron, 0=Air)')

# 7. Performance comparison
ax6 = plt.subplot(2, 3, 6)
initial_perf = performance_steps[0]
final_perf = performance_steps[-1]

metrics = ['Torque (Nm)', 'Efficiency', 'Torque Ripple']
initial_values = [initial_perf['torque'], initial_perf['efficiency'], initial_perf['torque_ripple']]
final_values = [final_perf['torque'], final_perf['efficiency'], final_perf['torque_ripple']]

x = np.arange(len(metrics))
width = 0.35

ax6.bar(x - width/2, initial_values, width, label='Initial', alpha=0.8)
ax6.bar(x + width/2, final_values, width, label='Optimized', alpha=0.8)

ax6.set_xlabel('Performance Metrics')
ax6.set_ylabel('Value')
ax6.set_title('Performance Comparison')
ax6.set_xticks(x)
ax6.set_xticklabels(metrics)
ax6.legend()
ax6.grid(True, alpha=0.3)

plt.suptitle('SynRM Topology Optimization Environment', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Print summary
print("\nSynRM Optimization Summary:")
print(f"Initial Performance: Torque = {initial_perf['torque']:.3f} Nm, Efficiency = {initial_perf['efficiency']:.3f}")
print(f"Final Performance: Torque = {final_perf['torque']:.3f} Nm, Efficiency = {final_perf['efficiency']:.3f}")
print(f"Torque Improvement: {(final_perf['torque'] - initial_perf['torque']):.3f} Nm ({100*(final_perf['torque']/initial_perf['torque'] - 1):.1f}%)")
print(f"Total Reward: {sum(rewards):.3f}")

display(MMarkdown("""
**Figure: SynRM Topology Optimization Environment**

This demonstration showcases the SynRM environment for reinforcement learning:

1. **Initial Solid Rotor**: Starting point with all-iron rotor (no flux barriers)
2. **Optimized Rotor Design**: Final topology with strategic flux barrier placement
3. **Performance Evolution**: Progressive improvement in torque and efficiency
4. **Reward Progression**: Agent receives positive feedback for performance improvements
5. **Final Topology**: Grid view showing material distribution (black = air barriers, white = iron)
6. **Performance Comparison**: Quantitative improvement in key metrics

**Key Design Insights:**
- **Radial flux barriers** (vertical patterns) improve torque production
- **Iron bridges** maintain structural integrity
- **Optimal air ratio** balances flux guidance and rotor strength
- **Progressive improvement** through sequential design modifications

This simplified environment captures the essential physics of SynRM design while remaining computationally tractable for reinforcement learning demonstrations.
"""))

# Add the missing polar visualization method
def visualize_rotor_polar(self, ax, topology=None, title="Rotor Topology"):
    """Visualize rotor topology in polar coordinates"""
    if topology is None:
        topology = self.topology
    
    # Draw rotor boundaries
    theta = np.linspace(0, 2*np.pi, 100)
    outer_r = self.rotor_outer_dia/2
    inner_r = self.rotor_inner_dia/2
    
    ax.plot(theta, [outer_r]*100, 'k-', linewidth=2)
    ax.plot(theta, [inner_r]*100, 'k-', linewidth=2)
    
    # Draw topology (show 4 quadrants)
    grid_extent = (self.rotor_outer_dia - self.rotor_inner_dia) / 2
    cell_size_r = grid_extent / self.grid_size
    cell_size_theta = (np.pi / 2) / self.grid_size  # Quarter circle per quadrant
    
    for quadrant in range(4):
        theta_offset = quadrant * np.pi / 2
        
        for i in range(self.grid_size):
            for j in range(self.grid_size):
                if topology[i, j] == 1:  # Iron
                    r_center = inner_r + (i + 0.5) * cell_size_r
                    theta_center = theta_offset + (j + 0.5) * cell_size_theta
                    
                    # Draw cell as a small arc segment
                    r_inner = r_center - cell_size_r/2
                    r_outer = r_center + cell_size_r/2
                    theta_start = theta_center - cell_size_theta/2
                    theta_end = theta_center + cell_size_theta/2
                    
                    theta_arc = np.linspace(theta_start, theta_end, 20)
                    for r in np.linspace(r_inner, r_outer, 3):
                        ax.plot(theta_arc, [r]*20, 'gray', linewidth=2, alpha=0.7)
    
    ax.set_ylim(0, outer_r + 10)
    ax.set_title(title)
    ax.grid(True, alpha=0.3)

# Add method to class
SynRMEnvironment.visualize_rotor_polar = visualize_rotor_polar
```````
