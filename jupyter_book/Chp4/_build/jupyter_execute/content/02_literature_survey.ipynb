### 1.2 Density-Based Methods

The Solid Isotropic Material with Penalization (SIMP) method has been widely used in topology optimization.

```````python
def simp_function(x, penal=3):
    """SIMP material interpolation function"""
    return x**penal

# Visualize SIMP penalization
x = np.linspace(0, 1, 100)
penalties = [1, 2, 3, 4, 5]

plt.figure(figsize=(10, 6))
for p in penalties:
    y = simp_function(x, p)
    plt.plot(x, y, label=f'Penalization = {p}', linewidth=2)

plt.xlabel('Density (x)')
plt.ylabel('Modified Density')
plt.title('SIMP Material Interpolation Function')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```````
