Inventors have always dreamed of machines that can think. Building intelligent machines involves the capability to learn from examples, experiences, or even from other machines [[Turing 1950]](#references).

This is significantly different from the supervised learning algorithms discussed in earlier chapters, where the task was well defined in terms of the labeled data, and the feedback provided to the learning algorithm was in the form of a correct response (ground truth/labeled data).

## 1.2 Reinforcement Learning: Learning from Experience

In contrast to supervised learning, a far more complicated task is to enable a learning technique to improve over time in an interactive environment through trial and error and using feedback from its actions and experiences. This type of learning, when coupled with neural networks, is the basis of **Deep Reinforcement Learning (DRL)**.

### 1.2.1 Psychological Foundations

To some extent, the psychology principle - 'Law of Effects' [[Thorndike 1913]](#references), associated with the behaviour of animals, can be related to the foundation of Reinforcement Learning. The notion states that:

> "responses to experiences that produce a satisfying situation become more likely to occur again in that situation, and responses that produce a discomforting effect become less likely" [[Gray 2011]](#references)

This principle was advanced by Edward Thorndike, who was the first to introduce the concept of 'reinforcement' in the area of the psychological principle of learning [[Thorndike 1898]](#references). The terms 'satisfying' and 'dissatisfying' were later replaced with the words 'reinforcing' and 'punishing' respectively [[Mazur 2015]](#references).

```python
import numpy as np
import matplotlib.pyplot as plt
from typing import List, Tuple, Dict, Optional

class SimpleGridWorld:
    """Simple grid world for demonstrating RL concepts"""
    
    def __init__(self, size: int = 5):
        self.size = size
        self.grid = np.zeros((size, size))
        self.goal = (size-1, size-1)
        self.obstacles = [(2, 2), (3, 3)]
        
        for obs in self.obstacles:
            self.grid[obs] = -1
        self.grid[self.goal] = 10
        
        self.reset()
    
    def reset(self) -> Tuple[int, int]:
        """Reset environment and return initial state"""
        self.current_pos = (0, 0)
        self.grid[self.current_pos] = 0
        return self.current_pos
    
    def step(self, action: int) -> Tuple[Tuple[int, int], float, bool, Dict]:
        """Take action and return (next_state, reward, done, info)"""
        # Actions: 0=up, 1=right, 2=down, 3=left
        moves = [(-1, 0), (0, 1), (1, 0), (0, -1)]
        
        # Calculate new position
        new_pos = (
            self.current_pos[0] + moves[action][0],
            self.current_pos[1] + moves[action][1]
        )
        
        # Check boundaries
        if (0 <= new_pos[0] < self.size and 
            0 <= new_pos[1] < self.size):
            
            # Check for obstacles
            if new_pos in self.obstacles:
                reward = -5.0
                done = False
            elif new_pos == self.goal:
                reward = 10.0
                done = True
            else:
                reward = -0.1  # Small negative reward for each step
                done = False
            
            # Update position
            self.grid[self.current_pos] = 0
            self.current_pos = new_pos
            self.grid[self.current_pos] = 1
        else:
            reward = -1.0  # Penalty for hitting boundary
            done = False
        
        return new_pos, reward, done, {'grid': self.grid.copy()}
    
    def render(self):
        """Visualize the grid world"""
        plt.figure(figsize=(6, 6))
        plt.imshow(self.grid, cmap='RdYlGn', vmin=-1, vmax=10)
        plt.colorbar(label='Reward')
        plt.title(f'Simple Grid World (Current: {self.current_pos})')
        plt.show()

# Demonstrate the simple RL environment
env = SimpleGridWorld()
state = env.reset()
print(f"Initial state: {state}")

# Take a few random actions
for step in range(10):
    action = np.random.randint(0, 4)  # Random action
    next_state, reward, done, info = env.step(action)
    print(f"Step {step + 1}: Action={action}, State={next_state}, Reward={reward:.2f}, Done={done}")
    if done:
        break

env.render()
```

## 1.3 DRL Breakthrough Applications

Deep Reinforcement Learning is expected to revolutionize the field of AI and represents a step towards building autonomous systems with a higher-level understanding of the physical world. The field received a lot of attention for several breakthrough developments:

## 1.4 DRL for Topology Optimization

While the present achievements of DRL in the field of robotics and computer games are impressive and required a massive amount of resources to discover, **this chapter extends the work in the field of DRL to Topology Optimization**.

### 1.4.1 The Topology Optimization Challenge

Topology optimization is the process of finding the optimal distribution of material within a design domain to achieve specific performance criteria. In electrical machine design, this involves:
- Maximizing performance metrics (torque, force, efficiency)
- Satisfying constraints (volume, manufacturability)
- Navigating complex design spaces
- Dealing with multi-objective trade-offs

### 1.4.2 Why DRL for Topology Optimization?

Traditional approaches to topology optimization have several limitations:

1. **Supervised Learning Limitations**: 
   - Require labeled optimal designs as training data
   - Limited generalization to unseen geometries
   - Collection bias in training datasets
   - Poor performance on novel design spaces

2. **Optimization Algorithm Limitations**:
   - Genetic algorithms require many function evaluations
   - Gradient-based methods struggle with discrete design variables
   - Limited adaptability to changing constraints

3. **Computational Cost**:
   - High-fidelity simulations are expensive
   - Optimization requires many iterations
   - Limited exploration of design space

### 1.4.3 DRL Advantages

Deep Reinforcement Learning addresses these limitations by:

- **Sequential Decision Making**: Learns through interaction with the environment
- **Adaptive Learning**: Can adapt to new geometries and constraints
- **Generalization**: Can transfer learning to similar optimization problems
- **Exploration-Exploitation Balance**: Systematically explores design space while exploiting promising regions
- **Online Learning**: Continuously improves with experience
