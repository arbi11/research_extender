# Deep Q-Learning: DQN Algorithm and C-Core Case Study

## Overview

This section introduces Deep Q-Networks (DQN) and demonstrates their application to C-core electromagnetic actuator topology optimization.

## Deep Q-Networks (DQN)

### Key Components

#### Neural Network Architecture
- Input: State representation (topology configuration)
- Hidden layers: Convolutional neural networks
- Output: Q-values for all possible actions

#### Experience Replay
- Stores transitions (state, action, reward, next_state)
- Random sampling for training
- Breaks correlation in training data

#### Target Network
- Stable Q-value estimation
- Periodic parameter updates
- Reduces moving target problem

## Algorithm Details

### Training Process
1. Initialize Q-network and target network
2. For each episode:
   - Observe current state
   - Select action using ε-greedy policy
   - Execute action and observe reward
   - Store transition in replay buffer
   - Sample random batch for training
   - Update Q-network parameters
   - Periodically update target network

### Loss Function
L(θ) = E[(r + γ max_a' Q(s', a'; θ-) - Q(s, a; θ))²]

where:
- θ: Current network parameters
- θ-: Target network parameters
- γ: Discount factor

## C-Core Case Study

### Problem Definition
- **Goal**: Optimize C-core electromagnetic actuator topology
- **State**: Current material distribution
- **Actions**: Add/remove material at specific locations
- **Reward**: Electromagnetic force generation

### State Representation
- 2D grid representation of topology
- Material presence/absence at each location
- Additional features (flux density, gradients)

### Action Space
- Discrete actions for material modification
- Coordinate-based selection
- Valid action masking

### Reward Design
- Primary: Electromagnetic force
- Secondary: Material usage efficiency
- Penalties: Constraint violations

## Implementation Considerations

### Network Architecture
- CNN for spatial feature extraction
- Fully connected layers for Q-value estimation
- ReLU activations for non-linearity

### Hyperparameters
- Learning rate: 1e-4 to 1e-3
- Discount factor: 0.99
- Experience replay buffer: 10000+ transitions
- Target network update: Every 1000 steps

## Performance Evaluation

### Metrics
- Final electromagnetic force
- Convergence speed
- Material efficiency
- Training stability

### Results
DQN demonstrates:
- Significant performance improvement over random search
- Effective learning of topology patterns
- Stable convergence behavior
- Good generalization to similar problems

## Advantages and Limitations

### Advantages
- Handles high-dimensional state spaces
- Learns complex spatial patterns
- Stable training with experience replay
- Good sample efficiency

### Limitations
- Discrete action space limitations
- Overestimation bias in Q-values
- Exploration challenges in large spaces
- Computational requirements

## Extensions and Improvements

- Double DQN: Reduces overestimation bias
- Dueling DQN: Separate state and advantage values
- Prioritized experience replay: Important transitions
- Distributional DQN: Value distribution learning

## Conclusion

Deep Q-Networks provide a powerful framework for topology optimization, effectively learning optimal design policies through interaction with simulated environments. The C-core case study demonstrates the practical applicability of DQN to electromagnetic design problems.
