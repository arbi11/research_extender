# Enhanced MDP: SeqTO-v2 Environment Modifications

## Overview

This section presents the SeqTO-v2 (Sequential Topology Optimization version 2) environment, an enhanced Markov Decision Process framework specifically designed for electromagnetic topology optimization.

## SeqTO-v2 Architecture

### Core Components

#### State Representation
- **Topology Grid**: 2D/3D material distribution
- **Physical Fields**: Flux density, field gradients
- **History**: Previous actions and their effects
- **Constraints**: Material limits, boundary conditions

#### Action Space
- **Material Placement**: Add material at specific locations
- **Material Removal**: Remove material from locations
- **Batch Operations**: Multiple simultaneous modifications
- **Symmetry Actions**: Enforce design symmetries

#### Reward Function
R(s,a,s') = α·F_performance + β·E_efficiency + γ·S_smoothness + δ·C_constraints

where:
- F_performance: Electromagnetic force/torque
- E_efficiency: Material usage efficiency
- S_smoothness: Topology smoothness
- C_constraints: Constraint satisfaction
