# 2. Literature Survey: Machine Learning for Topology Optimization

## 2.1 Introduction

This section surveys the application of machine learning techniques to topology optimization problems, with particular focus on supervised learning approaches and their limitations. Understanding these prior works motivates the deep reinforcement learning approach developed in this chapter.

## 2.2 Supervised Learning Approaches

### 2.2.1 Convolutional Neural Networks (CNNs)

Several researchers have explored using CNNs to predict optimal topologies:

**Direct Topology Prediction**:
- Train CNN on database of optimal designs
- Input: Boundary conditions, loads, constraints
- Output: Material distribution (topology)
- Fast inference once trained

**Limitations**:
1. Requires large dataset of optimal solutions
2. Dataset collection is expensive (must solve many TO problems)
3. Poor generalization to unseen boundary conditions
4. Not a true optimization algorithm (pattern matching)

### 2.2.2 Generative Adversarial Networks (GANs)

GANs have been applied to generate novel topologies:

**Approach**:
- Generator creates topology candidates
- Discriminator evaluates topology quality
- Adversarial training improves both networks

**Benefits**:
- Can generate diverse topologies
- Learns implicit design rules

**Limitations**:
1. Training instability
2. Mode collapse (generates similar designs)
3. No guarantee of optimality
4. Requires labeled data for discriminator

### 2.2.3 Autoencoders for Topology Representation

**Latent Space Learning**:
- Encoder: Topology → Low-dimensional representation
- Decoder: Representation → Topology
- Optimize in latent space (lower dimensional)

**Advantages**:
- Reduces dimensionality of optimization problem
- Smooth latent space for gradient-based optimization
- Can incorporate physics constraints in latent space

**Limitations**:
1. Reconstruction errors
2. Limited to topologies in training distribution
3. Still requires labeled optimal designs

## 2.3 Hybrid Approaches: DL + Traditional Optimization

### 2.3.1 Surrogate Models

Use neural networks as fast approximations of expensive simulations:

**Process**:
1. Train neural network to predict performance metrics
2. Use NN surrogate in optimization loop
3. Periodically validate with full FEA
4. Update surrogate with new data

**Benefits**:
- Reduces FEA evaluations (10-100x speedup)
- Can use with any optimization algorithm
- Adaptive improvement during optimization

**Challenges**:
- Surrogate accuracy critical
- Extrapolation errors
- Requires initial training data

### 2.3.2 Multi-Fidelity Optimization

Combine low-fidelity (fast, approximate) and high-fidelity (slow, accurate) models:

- Use ML to correct low-fidelity predictions
- Adaptively select which fidelity to use
- Balance accuracy vs computational cost

## 2.4 Fundamental Limitations of Supervised Learning for TO

### 2.4.1 Data Collection Bias

**Problem**: Training data comes from existing optimization algorithms
- Can only learn what optimizer discovered
- Inherits biases and limitations of data generation method
- Cannot discover better solutions than training data

**Example**: If training data generated by genetic algorithm with specific parameters, ML model learns GA biases.

### 2.4.2 Lack of Generalizability

Supervised models struggle with:
- New boundary conditions not in training set
- Different material properties
- Modified constraints
- Scaled geometries

**Root Cause**: Learn correlation, not causation. Don't understand underlying physics.

### 2.4.3 Not True Optimization

Supervised learning is **pattern matching**, not **optimization**:

- Predicts what optimizer would output
- Doesn't explore design space
- Cannot adapt to changing objectives
- No improvement during inference

## 2.5 Why Deep Reinforcement Learning?

DRL addresses the fundamental limitations:

### 2.5.1 No Labeled Data Required

- Learns through interaction with environment
- Generates own training data via exploration
- Can discover solutions not seen before

### 2.5.2 True Optimization

- Maximizes reward (performance metric)
- Explores design space systematically
- Improves policy over time
- Adapts to new problems

### 2.5.3 Generalization Through Transfer Learning

- Learn general electromagnetic principles
- Transfer knowledge to related problems
- Fine-tune on new constraints
- Multi-task learning across problem variations

### 2.5.4 Sequential Decision Making

TO is naturally sequential:
- Build topology incrementally
- Each decision affects future options
- MDP framework models this perfectly

## 2.6 Research Gaps Addressed

This chapter addresses the following gaps in existing literature:

1. **Scaling RL to High-Dimensional States**: Use function approximation (neural networks) to handle complex state spaces

2. **Physics-Informed State Representations**: Incorporate magnetic field distributions in state representation

3. **Generalization Across Problem Variations**: Demonstrate transfer learning across different excitation levels

4. **Computational Efficiency**: Show that DRL can be more efficient than traditional methods for families of related problems

5. **Empirical Validation**: Comprehensive comparison with genetic algorithms on realistic electromagnetic problems

## 2.7 Summary

The literature review reveals that:

- **Supervised learning** approaches have been explored but face fundamental limitations
- **Hybrid methods** show promise but still rely on labeled optimal designs
- **Deep RL** addresses key limitations by learning through interaction without labeled data
- **Research gaps** exist in scaling RL to complex TO problems with high-dimensional states

The following sections develop a deep reinforcement learning framework that addresses these limitations and demonstrates its effectiveness on electromagnetic topology optimization.
