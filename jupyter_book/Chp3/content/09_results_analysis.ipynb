## 9.1 Performance Metrics for Motor Prediction

### 9.1.1 Regression Metrics

Motor performance prediction requires careful selection of appropriate metrics that capture different aspects of model performance.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.metrics import mean_absolute_percentage_error, explained_variance_score
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Set random seeds for reproducibility
np.random.seed(42)

print("📊 Results Analysis and Visualization for Motor Performance Prediction")
print("=" * 70)
print("🎯 Chapter Goals:")
print("  • Master comprehensive performance metrics")
print("  • Learn error analysis techniques")
print("  • Create effective performance visualizations")
print("  • Understand comparative analysis methods")
print("  • Develop communication strategies")
```

```python
def calculate_comprehensive_metrics(y_true, y_pred, metric_names=None):
    """Calculate comprehensive performance metrics for motor prediction"""
    
    if metric_names is None:
        metric_names = ['MSE', 'RMSE', 'MAE', 'MAPE', 'R²', 'EVS', 'NSE']
    
    metrics = {}
    
    # Basic metrics
    if 'MSE' in metric_names:
        metrics['MSE'] = mean_squared_error(y_true, y_pred)
    
    if 'RMSE' in metric_names:
        metrics['RMSE'] = np.sqrt(mean_squared_error(y_true, y_pred))
    
    if 'MAE' in metric_names:
        metrics['MAE'] = mean_absolute_error(y_true, y_pred)
    
    if 'MAPE' in metric_names:
        metrics['MAPE'] = mean_absolute_percentage_error(y_true, y_pred) * 100
    
    if 'R²' in metric_names:
        metrics['R²'] = r2_score(y_true, y_pred)
    
    if 'EVS' in metric_names:
        metrics['EVS'] = explained_variance_score(y_true, y_pred)
    
    # Nash-Sutcliffe Efficiency (common in hydrology/motor performance)
    if 'NSE' in metric_names:
        numerator = np.sum((y_true - y_pred) ** 2)
        denominator = np.sum((y_true - np.mean(y_true)) ** 2)
        metrics['NSE'] = 1 - numerator / denominator
    
    # Additional motor-specific metrics
    if 'Normalized_RMSE' in metric_names:
        y_range = np.max(y_true) - np.min(y_true)
        metrics['Normalized_RMSE'] = np.sqrt(mean_squared_error(y_true, y_pred)) / y_range
    
    if 'Bias' in metric_names:
        metrics['Bias'] = np.mean(y_pred - y_true)
    
    if 'Max_Error' in metric_names:
        metrics['Max_Error'] = np.max(np.abs(y_pred - y_true))
    
    return metrics

def demonstrate_performance_metrics():
    """Demonstrate comprehensive performance metrics analysis"""
    
    # Generate synthetic motor performance predictions
    print("🔄 Generating Motor Performance Prediction Results...")
    
    np.random.seed(42)
    n_samples = 500
    
    # True performance values
    true_efficiency = 0.85 + 0.1 * np.random.rand(n_samples) - 0.05 * (np.random.rand(n_samples) - 0.5)**2
    true_power_loss = 10 + 15 * np.random.rand(n_samples) + 5 * np.sin(np.random.rand(n_samples) * 2 * np.pi)
    true_thermal_rise = 20 + 30 * np.random.rand(n_samples) + 10 * np.cos(np.random.rand(n_samples) * np.pi)
    
    # Model predictions with different error characteristics
    # Model 1: Good overall performance
    pred1_efficiency = true_efficiency + np.random.normal(0, 0.01, n_samples)
    pred1_power_loss = true_power_loss + np.random.normal(0, 1.5, n_samples)
    pred1_thermal_rise = true_thermal_rise + np.random.normal(0, 3.0, n_samples)
    
    # Model 2: Higher bias but lower variance
    pred2_efficiency = true_efficiency - 0.02 + np.random.normal(0, 0.008, n_samples)
    pred2_power_loss = true_power_loss + 2.0 + np.random.normal(0, 1.2, n_samples)
    pred2_thermal_rise = true_thermal_rise + 5.0 + np.random.normal(0, 2.5, n_samples)
    
    # Model 3: Non-uniform errors
    pred3_efficiency = true_efficiency + np.random.normal(0, 0.015 * (1 + 0.5 * true_efficiency), n_samples)
    pred3_power_loss = true_power_loss + np.random.normal(0, 2.0 * (1 + 0.3 * true_power_loss / 25), n_samples)
    pred3_thermal_rise = true_thermal_rise + np.random.normal(0, 4.0 * (1 + 0.2 * true_thermal_rise / 50), n_samples)
    
    # Organize data
    models = {
        'Model 1 (Balanced)': {
            'efficiency': pred1_efficiency,
            'power_loss': pred1_power_loss,
            'thermal_rise': pred1_thermal_rise
        },
        'Model 2 (High Bias)': {
            'efficiency': pred2_efficiency,
            'power_loss': pred2_power_loss,
            'thermal_rise': pred2_thermal_rise
        },
        'Model 3 (Heteroscedastic)': {
            'efficiency': pred3_efficiency,
            'power_loss': pred3_power_loss,
            'thermal_rise': pred3_thermal_rise
        }
    }
    
    true_values = {
        'efficiency': true_efficiency,
        'power_loss': true_power_loss,
        'thermal_rise': true_thermal_rise
    }
    
    print(f"✅ Generated {n_samples} samples for 3 models and 3 metrics")
    
    # Calculate comprehensive metrics
    print("\n📈 Calculating Comprehensive Performance Metrics...")
    
    metrics_comparison = {}
    
    for model_name, model_preds in models.items():
        metrics_comparison[model_name] = {}
        
        for metric_name, true_vals in true_values.items():
            pred_vals = model_preds[metric_name]
            metrics = calculate_comprehensive_metrics(true_vals, pred_vals)
            metrics_comparison[model_name][metric_name] = metrics
    
    # Create visualization
    fig, axes = plt.subplots(3, 3, figsize=(18, 15))
    fig.suptitle('Comprehensive Performance Metrics Analysis', fontsize=16, fontweight='bold')
    
    metric_names = ['efficiency', 'power_loss', 'thermal_rise']
    metric_display_names = ['Efficiency', 'Power Loss (W)', 'Thermal Rise (°C)']
    
    # Plot 1-3: Scatter plots with regression lines
    for i, (metric, display_name) in enumerate(zip(metric_names, metric_display_names)):
        ax = axes[0, i]
        colors = ['blue', 'red', 'green']
        
        for j, (model_name, model_preds) in enumerate(models.items()):
            pred_vals = model_preds[metric]
            true_vals = true_values[metric]
            
            # Scatter plot
            ax.scatter(true_vals, pred_vals, alpha=0.6, s=20, 
                      color=colors[j], label=model_name)
            
            # Perfect prediction line
            min_val = min(np.min(true_vals), np.min(pred_vals))
            max_val = max(np.max(true_vals), np.max(pred_vals))
            ax.plot([min_val, max_val], [min_val, max_val], 'k--', alpha=0.5)
        
        ax.set_xlabel(f'True {display_name}', fontweight='bold')
        ax.set_ylabel(f'Predicted {display_name}', fontweight='bold')
        ax.set_title(f'{display_name} Predictions', fontweight='bold')
        ax.legend(fontsize=8)
        ax.grid(True, alpha=0.3)
    
    # Plot 4-6: Residual analysis
    for i, (metric, display_name) in enumerate(zip(metric_names, metric_display_names)):
        ax = axes[1, i]
        
        for j, (model_name, model_preds) in enumerate(models.items()):
            pred_vals = model_preds[metric]
            true_vals = true_values[metric]
            residuals = pred_vals - true_vals
            
            ax.scatter(true_vals, residuals, alpha=0.6, s=20,
                      color=colors[j], label=model_name)
            ax.axhline(y=0, color='black', linestyle='--', alpha=0.5)
        
        ax.set_xlabel(f'True {display_name}', fontweight='bold')
        ax.set_ylabel('Residuals', fontweight='bold')
        ax.set_title(f'{display_name} Residual Analysis', fontweight='bold')
        ax.legend(fontsize=8)
        ax.grid(True, alpha=0.3)
    
    # Plot 7-9: Metric comparison heatmaps
    metric_keys = ['MSE', 'RMSE', 'MAE', 'R²', 'NSE']
    
    for i, metric in enumerate(metric_names):
        ax = axes[2, i]
        
        # Create matrix for heatmap
        matrix_data = []
        for model_name in models.keys():
            row = []
            for metric_key in metric_keys:
                value = metrics_comparison[model_name][metric][metric_key]
                # Normalize for better visualization
                if metric_key in ['R²', 'NSE']:
                    # Higher is better, invert for consistent coloring
                    normalized = 1 - value
                else:
                    # Lower is better, normalize
                    if metric_key == 'MSE':
                        normalized = value / 0.01 if metric == 'efficiency' else value / 10
                    elif metric_key == 'RMSE':
                        normalized = value / 0.1 if metric == 'efficiency' else value / 5
                    elif metric_key == 'MAE':
                        normalized = value / 0.08 if metric == 'efficiency' else value / 8
                    else:
                        normalized = value
                
                row.append(normalized)
            matrix_data.append(row)
        
        # Create heatmap
        im = ax.imshow(matrix_data, cmap='RdYlBu_r', aspect='auto')
        
        # Set labels
        ax.set_xticks(range(len(metric_keys)))
        ax.set_xticklabels(metric_keys, rotation=45)
        ax.set_yticks(range(len(models)))
        ax.set_yticklabels(list(models.keys()))
        
        # Add values to cells
        for j in range(len(models)):
            for k in range(len(metric_keys)):
                original_value = metrics_comparison[list(models.keys())[j]][metric][metric_keys[k]]
                if metric_keys[k] in ['R²', 'NSE']:
                    text_color = 'white' if original_value < 0.5 else 'black'
                else:
                    text_color = 'white' if matrix_data[j][k] > 0.5 else 'black'
                
                text = f'{original_value:.3f}'
                if metric_keys[k] in ['MSE', 'RMSE', 'MAE']:
                    if metric == 'efficiency':
                        text = f'{original_value:.4f}'
                    else:
                        text = f'{original_value:.2f}'
                
                ax.text(k, j, text, ha='center', va='center', 
                       color=text_color, fontweight='bold', fontsize=8)
        
        ax.set_title(f'{display_name} Metrics', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
    
    # Print detailed metrics table
    print("\n📊 Detailed Performance Metrics:")
    print("=" * 80)
    
    for metric_name in metric_names:
        print(f"\n🔧 {metric_name.replace('_', ' ').title()}:")
        print("-" * 40)
        
        # Create table header
        header = "Model        ".ljust(15)
        for metric_key in ['MSE', 'RMSE', 'MAE', 'R²', 'NSE']:
            header += f"{metric_key.ljust(8)}"
        print(header)
        print("-" * len(header))
        
        # Print values for each model
        for model_name in models.keys():
            row = model_name.ljust(15)
            for metric_key in ['MSE', 'RMSE', 'MAE', 'R²', 'NSE']:
                value = metrics_comparison[model_name][metric_name][metric_key]
                if metric_key in ['MSE', 'RMSE', 'MAE']:
                    if metric_name == 'efficiency':
                        row += f"{value:.4f}  "
                    else:
                        row += f"{value:.2f}  "
                else:
                    row += f"{value:.3f}  "
            print(row)
    
    return metrics_comparison

# Run performance metrics demonstration
metrics_results = demonstrate_performance_metrics()
```

## 9.2 Error Analysis and Diagnostic Techniques

Comprehensive error analysis helps identify specific weaknesses in model performance and guides improvement efforts.

```python
def error_analysis_diagnostics(y_true, y_pred, feature_names=None, operating_conditions=None):
    """Perform comprehensive error analysis and diagnostics"""
    
    # Basic error statistics
    residuals = y_pred - y_true
    absolute_errors = np.abs(residuals)
    squared_errors = residuals**2
    
    diagnostics = {
        'residuals': residuals,
        'absolute_errors': absolute_errors,
        'squared_errors': squared_errors,
        'mean_residual': np.mean(residuals),
        'std_residual': np.std(residuals),
        'mean_absolute_error': np.mean(absolute_errors),
        'rmse': np.sqrt(np.mean(squared_errors)),
        'max_error': np.max(absolute_errors),
        'median_absolute_error': np.median(absolute_errors)
    }
    
    # Error distribution analysis
    diagnostics['error_skewness'] = stats.skew(residuals)
    diagnostics['error_kurtosis'] = stats.kurtosis(residuals)
    
    # Normality test
    _, p_value = stats.normaltest(residuals)
    diagnostics['normality_p_value'] = p_value
    diagnostics['is_normal'] = p_value > 0.05
    
    # Heteroscedasticity test (Breusch-Pagan-like)
    if operating_conditions is not None:
        # Test if error variance changes with operating conditions
        condition_corr = []
        for i in range(operating_conditions.shape[1]):
            corr = np.corrcoef(operating_conditions[:, i], absolute_errors)[0, 1]
            condition_corr.append(abs(corr))
        
        diagnostics['heteroscedasticity_correlation'] = condition_corr
        diagnostics['max_heteroscedasticity_corr'] = max(condition_corr)
    
    # Percentile-based error analysis
    percentiles = [10, 25, 50, 75, 90, 95, 99]
    diagnostics['error_percentiles'] = {}
    for p in percentiles:
        diagnostics['error_percentiles'][f'p{p}'] = np.percentile(absolute_errors, p)
    
    # Error clustering analysis
    high_error_threshold = np.percentile(absolute_errors, 90)
    high_error_mask = absolute_errors > high_error_threshold
    
    diagnostics['high_error_threshold'] = high_error_threshold
    diagnostics['high_error_count'] = np.sum(high_error_mask)
    diagnostics['high_error_percentage'] = np.mean(high_error_mask) * 100
    
    if operating_conditions is not None:
        diagnostics['high_error_conditions'] = operating_conditions[high_error_mask]
        
        # Analyze conditions associated with high errors
        if feature_names is not None:
            high_error_stats = {}
            for i, feature in enumerate(feature_names):
                high_error_vals = operating_conditions[high_error_mask, i]
                all_vals = operating_conditions[:, i]
                
                high_error_stats[feature] = {
                    'mean': np.mean(high_error_vals),
                    'std': np.std(high_error_vals),
                    'overall_mean': np.mean(all_vals),
                    'overall_std': np.std(all_vals),
                    'z_score': (np.mean(high_error_vals) - np.mean(all_vals)) / np.std(all_vals)
                }
            
            diagnostics['high_error_condition_stats'] = high_error_stats
    
    return diagnostics

def demonstrate_error_analysis():
    """Demonstrate comprehensive error analysis techniques"""
    
    # Generate synthetic data with known error patterns
    print("🔍 Generating Data for Error Analysis...")
    
    np.random.seed(42)
    n_samples = 800
    
    # Operating conditions
    speeds = np.random.uniform(1000, 6000, n_samples)
    torques = np.random.uniform(20, 200, n_samples)
    currents = np.random.uniform(10, 150, n_samples)
    temperatures = np.random.uniform(20, 80, n_samples)
    
    X = np.column_stack([speeds, torques, currents, temperatures])
    feature_names = ['Speed (RPM)', 'Torque (Nm)', 'Current (A)', 'Temperature (°C)']
    
    # True efficiency (complex nonlinear relationship)
    true_efficiency = (0.92 - 0.15 * (speeds/6000)**2 - 0.08 * (torques/200)**2 
                      - 0.05 * (currents/150)**2 - 0.03 * (temperatures-50)/30)
    true_efficiency = np.clip(true_efficiency, 0.6, 0.98)
    
    # Model predictions with systematic errors
    # Base prediction with heteroscedastic noise
    base_error = np.random.normal(0, 0.015, n_samples)
    
    # Speed-dependent error (worse at high speeds)
    speed_error = 0.02 * (speeds/6000)**2 * np.random.normal(0, 1, n_samples)
    
    # Torque-dependent error (worse at high torques)
    torque_error = 0.015 * (torques/200) * np.random.normal(0, 1, n_samples)
    
    # Temperature bias (systematic underestimation at high temperatures)
    temp_bias = -0.01 * (temperatures - 50) / 30
    
    # Combine all errors
    total_error = base_error + speed_error + torque_error + temp_bias
    predicted_efficiency = true_efficiency + total_error
    
    print(f"✅ Generated {n_samples} samples with systematic error patterns")
    
    # Perform error analysis
    print("\n📊 Performing Comprehensive Error Analysis...")
    diagnostics = error_analysis_diagnostics(
        true_efficiency, predicted_efficiency, 
        feature_names=feature_names, 
        operating_conditions=X
    )
    
    # Create comprehensive visualization
    fig, axes = plt.subplots(3, 4, figsize=(20, 15))
    fig.suptitle('Comprehensive Error Analysis and Diagnostics', fontsize=16, fontweight='bold')
    
    # Plot 1: Residual histogram
    ax1 = axes[0, 0]
    ax1.hist(diagnostics['residuals'], bins=30, alpha=0.7, color='skyblue', edgecolor='black')
    ax1.axvline(x=0, color='red', linestyle='--', linewidth=2)
    ax1.axvline(x=diagnostics['mean_residual'], color='green', linestyle='-', linewidth=2,
               label=f'Mean: {diagnostics["mean_residual"]:.4f}')
    ax1.set_xlabel('Residuals', fontweight='bold')
    ax1.set_ylabel('Frequency', fontweight='bold')
    ax1.set_title('Residual Distribution', fontweight='bold')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Plot 2: Q-Q plot for normality
    ax2 = axes[0, 1]
    stats.probplot(diagnostics['residuals'], dist="norm", plot=ax2)
    ax2.set_title('Q-Q Plot (Normality Check)', fontweight='bold')
    ax2.grid(True, alpha=0.3)
    
    # Add normality test result
    normality_text = f"Normality p-value: {diagnostics['normality_p_value']:.3f}\n"
    normality_text += f"Is normal: {diagnostics['is_normal']}"
    ax2.text(0.02, 0.98, normality_text, transform=ax2.transAxes,
            verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
    
    # Plot 3: Residuals vs predicted values
    ax3 = axes[0, 2]
    ax3.scatter(predicted_efficiency, diagnostics['residuals'], alpha=0.6, s=10)
    ax3.axhline(y=0, color='red', linestyle='--', alpha=0.7)
    ax3.set_xlabel('Predicted Values', fontweight='bold')
    ax3.set_ylabel('Residuals', fontweight='bold')
    ax3.set_title('Residuals vs Predicted', fontweight='bold')
    ax3.grid(True, alpha=0.3)
    
    # Plot 4: Error percentiles
    ax4 = axes[0, 3]
    percentiles = list(diagnostics['error_percentiles'].keys())
    values = list(diagnostics['error_percentiles'].values())
    
    ax4.bar(range(len(percentiles)), values, alpha=0.7, color='lightcoral')
    ax4.set_xlabel('Percentile', fontweight='bold')
    ax4.set_ylabel('Absolute Error', fontweight='bold')
    ax4.set_title('Error Percentiles', fontweight='bold')
    ax4.set_xticks(range(len(percentiles)))
    ax4.set_xticklabels([p[1:] for p in percentiles], rotation=45)
    ax4.grid(True, alpha=0.3, axis='y')
    
    # Plot 5-8: Residuals vs operating conditions
    for i, (feature, name) in enumerate(zip(X.T, feature_names)):
        if i >= 4:
            break
        ax = axes[1, i]
        
        ax.scatter(feature, diagnostics['residuals'], alpha=0.6, s=10)
        ax.axhline(y=0, color='red', linestyle='--', alpha=0.7)
        
        # Add trend line
        z = np.polyfit(feature, diagnostics['residuals'], 1)
        p = np.poly1d(z)
        ax.plot(feature, p(feature), "r--", alpha=0.8, linewidth=2)
        
        # Add correlation info
        corr = np.corrcoef(feature, diagnostics['residuals'])[0, 1]
        ax.text(0.02, 0.98, f'Corr: {corr:.3f}', transform=ax.transAxes,
                verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
        
        ax.set_xlabel(name, fontweight='bold')
        ax.set_ylabel('Residuals', fontweight='bold')
        ax.set_title(f'Residuals vs {name.split()[0]}', fontweight='bold')
        ax.grid(True, alpha=0.3)
    
    # Plot 9: Absolute errors vs operating conditions (heatmap style)
    ax9 = axes[2, 0]
    # Create 2D heatmap of errors vs speed and torque
    speed_bins = np.linspace(1000, 6000, 20)
    torque_bins = np.linspace(20, 200, 15)
    
    error_matrix = np.zeros((len(speed_bins)-1, len(torque_bins)-1))
    
    for i in range(len(speed_bins)-1):
        for j in range(len(torque_bins)-1):
            mask = ((speeds >= speed_bins[i]) & (speeds < speed_bins[i+1]) &
                   (torques >= torque_bins[j]) & (torques < torque_bins[j+1]))
            if np.sum(mask) > 0:
                error_matrix[i, j] = np.mean(diagnostics['absolute_errors'][mask])
            else:
                error_matrix[i, j] = np.nan
    
    im = ax9.imshow(error_matrix, aspect='auto', origin='lower', cmap='viridis')
    ax9.set_xlabel('Torque (Nm)', fontweight='bold')
    ax9.set_ylabel('Speed (RPM)', fontweight='bold')
    ax9.set_title('Error Heatmap (Speed vs Torque)', fontweight='bold')
    
    # Set ticks
    ax9.set_xticks(range(0, len(torque_bins), 3))
    ax9.set_xticklabels([f'{int(torque_bins[i])}' for i in range(0, len(torque_bins), 3)])
    ax9.set_yticks(range(0, len(speed_bins), 4))
    ax9.set_yticklabels([f'{int(speed_bins[i])}' for i in range(0, len(speed_bins), 4)])
    
    plt.colorbar(im, ax=ax9, label='Mean Absolute Error')
    
    # Plot 10: High error condition analysis
    ax10 = axes[2, 1]
    if 'high_error_condition_stats' in diagnostics:
        features = list(diagnostics['high_error_condition_stats'].keys())
        z_scores = [diagnostics['high_error_condition_stats'][f]['z_score'] 
                   for f in features]
        
        colors = ['red' if abs(z) > 1 else 'blue' for z in z_scores]
        bars = ax10.bar(range(len(features)), z_scores, alpha=0.7, color=colors)
        
        ax10.set_xlabel('Operating Condition', fontweight='bold')
        ax10.set_ylabel('Z-Score (High Error vs Overall)', fontweight='bold')
        ax10.set_title('High Error Conditions Analysis', fontweight='bold')
        ax10.set_xticks(range(len(features)))
        ax10.set_xticklabels([f.split()[0] for f in features], rotation=45)
        ax10.axhline(y=0, color='black', linestyle='-', alpha=0.5)
        ax10.axhline(y=1, color='red', linestyle='--', alpha=0.5, label='Significant')
        ax10.axhline(y=-1, color='red', linestyle='--', alpha=0.5)
        ax10.legend()
        ax10.grid(True, alpha=0.3, axis='y')
    
    # Plot 11: Error magnitude distribution
    ax11 = axes[2, 2]
    error_categories = ['Low (<25%)', 'Medium (25-75%)', 'High (75-90%)', 'Very High (>90%)']
    thresholds = [diagnostics['error_percentiles']['p25'], 
                  diagnostics['error_percentiles']['p75'],
                  diagnostics['error_percentiles']['p90']]
    
    category_counts = []
    
    # Low errors
    category_counts.append(np.sum(diagnostics['absolute_errors'] < thresholds[0]))
    # Medium errors
    category_counts.append(np.sum((diagnostics['absolute_errors'] >= thresholds[0]) & 
                                 (diagnostics['absolute_errors'] < thresholds[1])))
    # High errors
    category_counts.append(np.sum((diagnostics['absolute_errors'] >= thresholds[1]) & 
                                 (diagnostics['absolute_errors'] < thresholds[2])))
    # Very high errors
    category_counts.append(np.sum(diagnostics['absolute_errors'] >= thresholds[2]))
    
    colors = ['green', 'yellow', 'orange', 'red']
    wedges, texts, autotexts = ax11.pie(category_counts, labels=error_categories, 
                                         colors=colors, autopct='%1.1f%%', startangle=90)
    ax11.set_title('Error Magnitude Distribution', fontweight='bold')
    
    # Plot 12: Summary statistics table
    ax12 = axes[2, 3]
    ax12.axis('off')
    
    # Create summary statistics text
    summary_text = "Error Analysis Summary:\n\n"
    summary_text += f"Mean Residual: {diagnostics['mean_residual']:.4f}\n"
    summary_text += f"Std Residual: {diagnostics['std_residual']:.4f}\n"
    summary_text += f"RMSE: {diagnostics['rmse']:.4f}\n"
    summary_text += f"MAE: {diagnostics['mean_absolute_error']:.4f}\n"
    summary_text += f"Max Error: {diagnostics['max_error']:.4f}\n\n"
    summary_text += f"Skewness: {diagnostics['error_skewness']:.3f}\n"
    summary_text += f"Kurtosis: {diagnostics['error_kurtosis']:.3f}\n"
    summary_text += f"Normal: {diagnostics['is_normal']}\n\n"
    summary_text += f"High Error Rate: {diagnostics['high_error_percentage']:.1f}%\n"
    summary_text += f"Max Heteroscedasticity: {diagnostics['max_heteroscedasticity_corr']:.3f}"
    
    ax12.text(0.1, 0.9, summary_text, transform=ax12.transAxes,
             verticalalignment='top', fontfamily='monospace',
             bbox=dict(boxstyle='round', facecolor='lightgray', alpha=0.8))
    
    plt.tight_layout()
    plt.show()
    
    # Print detailed analysis
    print("\n📊 Error Analysis Results:")
    print("=" * 50)
    print(f"📈 Mean Residual: {diagnostics['mean_residual']:.6f} (bias indicator)")
    print(f"📊 RMSE: {diagnostics['rmse']:.6f}")
    print(f"🎯 MAE: {diagnostics['mean_absolute_error']:.6f}")
    print(f"⚠️ Max Error: {diagnostics['max_error']:.6f}")
    print(f"📉 Normal Distribution: {diagnostics['is_normal']} (p={diagnostics['normality_p_value']:.3f})")
    print(f"🔥 High Error Cases: {diagnostics['high_error_percentage']:.1f}% of data")
    print(f"🌡️ Max Heteroscedasticity: {diagnostics['max_heteroscedasticity_corr']:.3f}")
    
    if 'high_error_condition_stats' in diagnostics:
        print("\n🎯 High Error Conditions:")
        for feature, stats in diagnostics['high_error_condition_stats'].items():
            if abs(stats['z_score']) > 1:
                direction = "higher" if stats['z_score'] > 0 else "lower"
                print(f"  • {feature}: {direction} than average (z={stats['z_score']:.2f})")
    
    return diagnostics

# Run error analysis demonstration
error_diagnostics = demonstrate_error_analysis()
```

## 9.3 Motor Performance Map Visualization

Effective visualization of motor performance maps is crucial for communicating model results and identifying optimal operating regions.

```python
def create_performance_map_visualization(speed_range, torque_range, efficiency_map, 
                                        title="Motor Efficiency Map",
                                        contour_levels=15, cmap='viridis'):
    """Create comprehensive motor performance map visualization"""
    
    fig, axes = plt.subplots(2, 2, figsize=(15, 12))
    fig.suptitle(title, fontsize=16, fontweight='bold')
    
    # Plot 1: Main contour map
    ax1 = axes[0, 0]
    
    # Create contour plot
    contour = ax1.contourf(speed_range, torque_range, efficiency_map, 
                          levels=contour_levels, cmap=cmap, alpha=0.8)
    
    # Add contour lines
    contour_lines = ax1.contour(speed_range, torque_range, efficiency_map, 
                               levels=10, colors='black', linewidths=0.5, alpha=0.5)
    ax1.clabel(contour_lines, inline=True, fontsize=8, fmt='%.2f')
    
    # Find and mark optimum
    max_idx = np.unravel_index(np.argmax(efficiency_map), efficiency_map.shape)
    opt_speed = speed_range[max_idx[1], max_idx[0]]
    opt_torque = torque_range[max_idx[1], max_idx[0]]
    opt_efficiency = efficiency_map[max_idx[1], max_idx[0]]
    
    ax1.plot(opt_speed, opt_torque, 'r*', markersize=15, 
            label=f'Optimal: {opt_efficiency:.3f}')
    
    ax1.set_xlabel('Speed (RPM)', fontweight='bold')
    ax1.set_ylabel('Torque (Nm)', fontweight='bold')
    ax1.set_title('Efficiency Contour Map', fontweight='bold')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Add colorbar
    cbar1 = plt.colorbar(contour, ax=ax1)
    cbar1.set_label('Efficiency', fontweight='bold')
    
    # Plot 2: 3D surface plot
    ax2 = axes[0, 1] = fig.add_subplot(2, 2, 2, projection='3d')
    
    # Create surface plot
    surf = ax2.plot_surface(speed_range, torque_range, efficiency_map,
                           cmap=cmap, alpha=0.8, antialiased=True)
    
    ax2.set_xlabel('Speed (RPM)', fontweight='bold')
    ax2.set_ylabel('Torque (Nm)', fontweight='bold')
    ax2.set_zlabel('Efficiency', fontweight='bold')
    ax2.set_title('3D Efficiency Surface', fontweight='bold')
    
    # Add colorbar
    cbar2 = plt.colorbar(surf, ax=ax2, shrink=0.5, aspect=5)
    cbar2.set_label('Efficiency', fontweight='bold')
    
    # Plot 3: Efficiency regions
    ax3 = axes[1, 0]
    
    # Define efficiency regions
    high_eff_mask = efficiency_map >= 0.90
    med_eff_mask = (efficiency_map >= 0.85) & (efficiency_map < 0.90)
    low_eff_mask = efficiency_map < 0.85
    
    # Create custom colormap for regions
    region_map = np.zeros_like(efficiency_map)
    region_map[high_eff_mask] = 3
    region_map[med_eff_mask] = 2
    region_map[low_eff_mask] = 1
    
    im = ax3.imshow(region_map, extent=[speed_range.min(), speed_range.max(),
                                       torque_range.min(), torque_range.max()],
                     aspect='auto', origin='lower', cmap='RdYlGn', alpha=0.7)
    
    # Add region labels
    ax3.text(opt_speed, opt_torque, 'OPTIMAL', ha='center', va='center',
            fontweight='bold', bbox=dict(boxstyle='round', facecolor='red', alpha=0.7))
    
    ax3.set_xlabel('Speed (RPM)', fontweight='bold')
    ax3.set_ylabel('Torque (Nm)', fontweight='bold')
    ax3.set_title('Efficiency Operating Regions', fontweight='bold')
    ax3.grid(True, alpha=0.3)
    
    # Add custom legend
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor='red', alpha=0.7, label='Low (< 85%)'),
        Patch(facecolor='yellow', alpha=0.7, label='Medium (85-90%)'),
        Patch(facecolor='green', alpha=0.7, label='High (≥ 90%)')
    ]
    ax3.legend(handles=legend_elements, loc='upper right')
    
    # Plot 4: Performance statistics
    ax4 = axes[1, 1]
    
    # Calculate statistics
    eff_stats = {
        'Mean': np.mean(efficiency_map),
        'Std': np.std(efficiency_map),
        'Min': np.min(efficiency_map),
        'Max': np.max(efficiency_map),
        'Optimal': opt_efficiency
    }
    
    # Bar chart of statistics
    stats_names = list(eff_stats.keys())
    stats_values = list(eff_stats.values())
    colors = ['blue', 'red', 'orange', 'green', 'purple']
    
    bars = ax4.bar(stats_names, stats_values, alpha=0.7, color=colors)
    
    # Add value labels on bars
    for bar, value in zip(bars, stats_values):
        height = bar.get_height()
        ax4.text(bar.get_x() + bar.get_width()/2., height + 0.005,
                f'{value:.3f}', ha='center', va='bottom', fontweight='bold')
    
    ax4.set_ylabel('Efficiency Value', fontweight='bold')
    ax4.set_title('Performance Statistics', fontweight='bold')
    ax4.set_ylim(0, max(stats_values) * 1.15)
    ax4.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.show()
    
    return {
        'optimal_point': (opt_speed, opt_torque, opt_efficiency),
        'statistics': eff_stats,
        'regions': {
            'high_efficiency': np.sum(high_eff_mask) / high_eff_mask.size,
            'medium_efficiency': np.sum(med_eff_mask) / med_eff_mask.size,
            'low_efficiency': np.sum(low_eff_mask) / low_eff_mask.size
        }
    }

def demonstrate_performance_maps():
    """Demonstrate motor performance map visualization techniques"""
    
    # Generate synthetic motor performance data
    print("🗺️ Generating Motor Performance Maps...")
    
    # Create operating condition grid
    speeds = np.linspace(1000, 6000, 50)
    torques = np.linspace(20, 200, 40)
    speed_grid, torque_grid = np.meshgrid(speeds, torques)
    
    # Generate efficiency maps for different motor types
    # IPM Motor - more efficient at medium speeds
    ipm_efficiency = 0.92 - 0.15 * (speed_grid/6000)**2 - 0.08 * (torque_grid/200)**2
    ipm_efficiency += 0.03 * np.sin(speed_grid / 1500) * np.cos(torque_grid / 80)
    ipm_efficiency = np.clip(ipm_efficiency, 0.6, 0.98)
    
    # FSCW Motor - better at high speeds
    fscw_efficiency = 0.88 - 0.08 * (speed_grid/6000)**2 - 0.12 * (torque_grid/200)**2
    fscw_efficiency += 0.04 * (1 - (speed_grid/6000 - 0.5)**2)  # Peak at high speed
    fscw_efficiency = np.clip(fscw_efficiency, 0.6, 0.95)
    
    # Predicted efficiency (with some prediction error)
    np.random.seed(42)
    prediction_noise = np.random.normal(0, 0.02, speed_grid.shape)
    predicted_efficiency = ipm_efficiency + prediction_noise
    predicted_efficiency = np.clip(predicted_efficiency, 0.6, 0.98)
    
    # Prediction error map
    error_map = predicted_efficiency - ipm_efficiency
    
    print(f"✅ Generated performance maps:")
    print(f"  • IPM Motor: {ipm_efficiency.shape}")
    print(f"  • FSCW Motor: {fscw_efficiency.shape}")
    print(f"  • Predicted: {predicted_efficiency.shape}")
    print(f"  • Error Map: {error_map.shape}")
    
    # Visualize IPM motor performance
    print("\n📊 Visualizing IPM Motor Performance...")
    ipm_analysis = create_performance_map_visualization(
        speed_grid, torque_grid, ipm_efficiency,
        title="IPM Motor Efficiency Map",
        cmap='viridis'
    )
    
    # Visualize FSCW motor performance
    print("\n📊 Visualizing FSCW Motor Performance...")
    fscw_analysis = create_performance_map_visualization(
        speed_grid, torque_grid, fscw_efficiency,
        title="FSCW Motor Efficiency Map",
        cmap='plasma'
    )
    
    # Visualize prediction error
    print("\n📊 Visualizing Prediction Error Map...")
    
    fig, axes = plt.subplots(2, 2, figsize=(15, 12))
    fig.suptitle('Prediction Error Analysis', fontsize=16, fontweight='bold')
    
    # Plot 1: Error contour map
    ax1 = axes[0, 0]
    error_levels = np.linspace(np.min(error_map), np.max(error_map), 20)
    contour = ax1.contourf(speed_grid, torque_grid, error_map, 
                          levels=error_levels, cmap='RdBu_r', alpha=0.8)
    
    ax1.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    ax1.set_xlabel('Speed (RPM)', fontweight='bold')
    ax1.set_ylabel('Torque (Nm)', fontweight='bold')
    ax1.set_title('Prediction Error Map', fontweight='bold')
    plt.colorbar(contour, ax=ax1, label='Prediction Error')
    ax1.grid(True, alpha=0.3)
    
    # Plot 2: Absolute error map
    ax2 = axes[0, 1]
    abs_error = np.abs(error_map)
    contour_abs = ax2.contourf(speed_grid, torque_grid, abs_error,
                             levels=15, cmap='Reds', alpha=0.8)
    ax2.set_xlabel('Speed (RPM)', fontweight='bold')
    ax2.set_ylabel('Torque (Nm)', fontweight='bold')
    ax2.set_title('Absolute Prediction Error', fontweight='bold')
    plt.colorbar(contour_abs, ax=ax2, label='Absolute Error')
    ax2.grid(True, alpha=0.3)
    
    # Plot 3: Error statistics by speed
    ax3 = axes[1, 0]
    error_by_speed = np.mean(error_map, axis=0)
    std_by_speed = np.std(error_map, axis=0)
    
    ax3.plot(speeds, error_by_speed, 'b-', linewidth=2, label='Mean Error')
    ax3.fill_between(speeds, error_by_speed - std_by_speed, error_by_speed + std_by_speed,
                     alpha=0.3, color='blue', label='±1 Std')
    ax3.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    ax3.set_xlabel('Speed (RPM)', fontweight='bold')
    ax3.set_ylabel('Prediction Error', fontweight='bold')
    ax3.set_title('Error vs Speed', fontweight='bold')
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    # Plot 4: Error statistics by torque
    ax4 = axes[1, 1]
    error_by_torque = np.mean(error_map, axis=1)
    std_by_torque = np.std(error_map, axis=1)
    
    ax4.plot(torques, error_by_torque, 'r-', linewidth=2, label='Mean Error')
    ax4.fill_between(torques, error_by_torque - std_by_torque, error_by_torque + std_by_torque,
                     alpha=0.3, color='red', label='±1 Std')
    ax4.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    ax4.set_xlabel('Torque (Nm)', fontweight='bold')
    ax4.set_ylabel('Prediction Error', fontweight='bold')
    ax4.set_title('Error vs Torque', fontweight='bold')
    ax4.legend()
    ax4.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Print analysis summary
    print("\n📊 Performance Map Analysis Summary:")
    print("=" * 50)
    print(f"\n🏆 IPM Motor:")
    print(f"  Optimal Point: {ipm_analysis['optimal_point'][0]:.0f} RPM, {ipm_analysis['optimal_point'][1]:.0f} Nm")
    print(f"  Optimal Efficiency: {ipm_analysis['optimal_point'][2]:.3f}")
    print(f"  High Efficiency Region: {ipm_analysis['regions']['high_efficiency']*100:.1f}%")
    
    print(f"\n🏆 FSCW Motor:")
    print(f"  Optimal Point: {fscw_analysis['optimal_point'][0]:.0f} RPM, {fscw_analysis['optimal_point'][1]:.0f} Nm")
    print(f"  Optimal Efficiency: {fscw_analysis['optimal_point'][2]:.3f}")
    print(f"  High Efficiency Region: {fscw_analysis['regions']['high_efficiency']*100:.1f}%")
    
    print(f"\n🎯 Prediction Accuracy:")
    print(f"  Mean Error: {np.mean(error_map):.4f}")
    print(f"  Mean Absolute Error: {np.mean(np.abs(error_map)):.4f}")
    print(f"  Max Error: {np.max(np.abs(error_map)):.4f}")
    
    return {
        'ipm_analysis': ipm_analysis,
        'fscw_analysis': fscw_analysis,
        'error_map': error_map
    }

# Run performance map demonstration
performance_map_results = demonstrate_performance_maps()
```

## 9.4 Comparative Analysis and Model Selection

Comparative analysis helps select the best model for specific applications by evaluating multiple criteria and considering trade-offs between different performance aspects.

```python
def comparative_model_analysis(models_results, model_names, criteria_weights=None):
    """Perform comprehensive comparative analysis of multiple models"""
    
    if criteria_weights is None:
        criteria_weights = {
            'accuracy': 0.4,
            'robustness': 0.2,
            'efficiency': 0.2,
            'interpretability': 0.1,
            'complexity': 0.1
        }
    
    # Normalize criteria (higher is better for all)
    # For complexity, lower is better, so we invert it
    
    # Calculate weighted scores
    model_scores = {}
    
    for i, model_name in enumerate(model_names):
        results = models_results[i]
        
        # Accuracy score (based on R²)
        accuracy_score = np.mean([results['efficiency']['R²'], 
                                   results['power_loss']['R²'], 
                                   results['thermal_rise']['R²']])
        
        # Robustness score (inverse of error variance)
        robustness_score = 1 / (1 + np.std([results['efficiency']['RMSE'],
                                             results['power_loss']['RMSE'],
                                             results['thermal_rise']['RMSE']]))
        
        # Efficiency score (based on computational metrics)
        efficiency_score = 0.8  # Placeholder - would depend on actual timing/memory usage
        
        # Interpretability score (based on model type)
        interpretability_scores = {
            'Linear Regression': 1.0,
            'Random Forest': 0.7,
            'GRU': 0.5,
            'GRU + Attention': 0.3,
            'Ensemble': 0.4
        }
        interpretability_score = interpretability_scores.get(model_name, 0.5)
        
        # Complexity score (inverse of parameter count)
        complexity_scores = {
            'Linear Regression': 1.0,
            'Random Forest': 0.6,
            'GRU': 0.4,
            'GRU + Attention': 0.2,
            'Ensemble': 0.3
        }
        complexity_score = complexity_scores.get(model_name, 0.5)
        
        # Calculate weighted total score
        total_score = (
            criteria_weights['accuracy'] * accuracy_score +
            criteria_weights['robustness'] * robustness_score +
            criteria_weights['efficiency'] * efficiency_score +
            criteria_weights['interpretability'] * interpretability_score +
            criteria_weights['complexity'] * complexity_score
        )
        
        model_scores[model_name] = {
            'accuracy': accuracy_score,
            'robustness': robustness_score,
            'efficiency': efficiency_score,
            'interpretability': interpretability_score,
            'complexity': complexity_score,
            'total_score': total_score
        }
    
    return model_scores

def demonstrate_comparative_analysis():
    """Demonstrate comprehensive comparative model analysis"""
    
    # Simulate results from multiple models
    print("🔄 Simulating Results from Multiple Models...")
    
    model_names = ['Linear Regression', 'Random Forest', 'GRU', 'GRU + Attention', 'Ensemble']
    
    # Generate synthetic results for each model
    np.random.seed(42)
    models_results = []
    
    for i, model_name in enumerate(model_names):
        # Different performance characteristics for each model
        if model_name == 'Linear Regression':
            # Lower accuracy but high interpretability
            r2_efficiency = 0.75 + np.random.normal(0, 0.02)
            r2_power_loss = 0.70 + np.random.normal(0, 0.03)
            r2_thermal = 0.72 + np.random.normal(0, 0.025)
            rmse_efficiency = 0.035 + np.random.normal(0, 0.005)
            rmse_power_loss = 3.5 + np.random.normal(0, 0.5)
            rmse_thermal = 6.0 + np.random.normal(0, 0.8)
        
        elif model_name == 'Random Forest':
            # Good balance of accuracy and interpretability
            r2_efficiency = 0.85 + np.random.normal(0, 0.015)
            r2_power_loss = 0.82 + np.random.normal(0, 0.02)
            r2_thermal = 0.84 + np.random.normal(0, 0.018)
            rmse_efficiency = 0.025 + np.random.normal(0, 0.003)
            rmse_power_loss = 2.5 + np.random.normal(0, 0.3)
            rmse_thermal = 4.5 + np.random.normal(0, 0.5)
        
        elif model_name == 'GRU':
            # High accuracy for sequential data
            r2_efficiency = 0.90 + np.random.normal(0, 0.01)
            r2_power_loss = 0.88 + np.random.normal(0, 0.015)
            r2_thermal = 0.89 + np.random.normal(0, 0.012)
            rmse_efficiency = 0.018 + np.random.normal(0, 0.002)
            rmse_power_loss = 1.8 + np.random.normal(0, 0.2)
            rmse_thermal = 3.2 + np.random.normal(0, 0.3)
        
        elif model_name == 'GRU + Attention':
            # Highest accuracy but complex
            r2_efficiency = 0.93 + np.random.normal(0, 0.008)
            r2_power_loss = 0.91 + np.random.normal(0, 0.01)
            r2_thermal = 0.92 + np.random.normal(0, 0.009)
            rmse_efficiency = 0.015 + np.random.normal(0, 0.0015)
            rmse_power_loss = 1.5 + np.random.normal(0, 0.15)
            rmse_thermal = 2.8 + np.random.normal(0, 0.25)
        
        else:  # Ensemble
            # Good robustness, moderate complexity
            r2_efficiency = 0.91 + np.random.normal(0, 0.009)
            r2_power_loss = 0.89 + np.random.normal(0, 0.012)
            r2_thermal = 0.90 + np.random.normal(0, 0.011)
            rmse_efficiency = 0.016 + np.random.normal(0, 0.0018)
            rmse_power_loss = 1.6 + np.random.normal(0, 0.18)
            rmse_thermal = 3.0 + np.random.normal(0, 0.28)
        
        model_results = {
            'efficiency': {
                'R²': max(0, min(1, r2_efficiency)),
                'RMSE': max(0, rmse_efficiency)
            },
            'power_loss': {
                'R²': max(0, min(1, r2_power_loss)),
                'RMSE': max(0, rmse_power_loss)
            },
            'thermal_rise': {
                'R²': max(0, min(1, r2_thermal)),
                'RMSE': max(0, rmse_thermal)
            }
        }
        
        models_results.append(model_results)
    
    print(f"✅ Generated results for {len(model_names)} models")
    
    # Perform comparative analysis
    print("\n📊 Performing Comparative Analysis...")
    
    # Standard analysis weights
    standard_weights = {
        'accuracy': 0.4,
        'robustness': 0.2,
        'efficiency': 0.2,
        'interpretability': 0.1,
        'complexity': 0.1
    }
    
    # Research-focused weights
    research_weights = {
        'accuracy': 0.6,
        'robustness': 0.2,
        'efficiency': 0.1,
        'interpretability': 0.05,
        'complexity': 0.05
    }
    
    # Industry-focused weights
    industry_weights = {
        'accuracy': 0.3,
        'robustness': 0.3,
        'efficiency': 0.25,
        'interpretability': 0.1,
        'complexity': 0.05
    }
    
    standard_scores = comparative_model_analysis(models_results, model_names, standard_weights)
    research_scores = comparative_model_analysis(models_results, model_names, research_weights)
    industry_scores = comparative_model_analysis(models_results, model_names, industry_weights)
    
    # Create visualization
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    fig.suptitle('Comprehensive Model Comparison Analysis', fontsize=16, fontweight='bold')
    
    # Plot 1: Accuracy comparison (R²)
    ax1 = axes[0, 0]
    accuracy_scores = [np.mean([models_results[i]['efficiency']['R²'],
                                models_results[i]['power_loss']['R²'],
                                models_results[i]['thermal_rise']['R²']]) for i in range(len(model_names))]
    
    bars = ax1.bar(model_names, accuracy_scores, alpha=0.7, color='skyblue')
    ax1.set_ylabel('Average R²', fontweight='bold')
    ax1.set_title('Model Accuracy Comparison', fontweight='bold')
    ax1.set_ylim(0, 1)
    ax1.tick_params(axis='x', rotation=45)
    ax1.grid(True, alpha=0.3, axis='y')
    
    # Add value labels
    for bar, score in zip(bars, accuracy_scores):
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{score:.3f}', ha='center', va='bottom', fontweight='bold')
    
    # Plot 2: Robustness comparison (inverse of RMSE variance)
    ax2 = axes[0, 1]
    robustness_scores = [standard_scores[name]['robustness'] for name in model_names]
    
    bars = ax2.bar(model_names, robustness_scores, alpha=0.7, color='lightcoral')
    ax2.set_ylabel('Robustness Score', fontweight='bold')
    ax2.set_title('Model Robustness Comparison', fontweight='bold')
    ax2.tick_params(axis='x', rotation=45)
    ax2.grid(True, alpha=0.3, axis='y')
    
    # Add value labels
    for bar, score in zip(bars, robustness_scores):
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{score:.3f}', ha='center', va='bottom', fontweight='bold')
    
    # Plot 3: Multi-criteria radar chart
    ax3 = axes[0, 2]
    categories = ['Accuracy', 'Robustness', 'Efficiency', 'Interpretability', 'Complexity']
    
    # Select top 3 models for radar chart
    top_models = sorted(model_names, 
                      key=lambda x: standard_scores[x]['total_score'], 
                      reverse=True)[:3]
    
    # Create radar chart
    angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()
    angles += angles[:1]  # Complete the circle
    
    ax3.remove()
    ax3 = fig.add_subplot(2, 3, 3, projection='polar')
    
    colors = ['blue', 'red', 'green']
    for i, model_name in enumerate(top_models):
        values = [standard_scores[model_name][cat.lower()] for cat in categories]
        values += values[:1]  # Complete the circle
        
        ax3.plot(angles, values, 'o-', linewidth=2, label=model_name, color=colors[i])
        ax3.fill(angles, values, alpha=0.25, color=colors[i])
    
    ax3.set_xticks(angles[:-1])
    ax3.set_xticklabels(categories)
    ax3.set_ylim(0, 1)
    ax3.set_title('Multi-Criteria Comparison\n(Top 3 Models)', fontweight='bold', pad=20)
    ax3.legend(loc='upper right', bbox_to_anchor=(1.3, 1.0))
    
    # Plot 4: Application scenario comparison
    ax4 = axes[1, 0]
    
    scenarios = ['Standard', 'Research', 'Industry']
    scenario_scores = {
        'Standard': [standard_scores[name]['total_score'] for name in model_names],
        'Research': [research_scores[name]['total_score'] for name in model_names],
        'Industry': [industry_scores[name]['total_score'] for name in model_names]
    }
    
    x = np.arange(len(model_names))
    width = 0.25
    
    bars1 = ax4.bar(x - width, scenario_scores['Standard'], width, 
                    label='Standard', color='skyblue', alpha=0.7)
    bars2 = ax4.bar(x, scenario_scores['Research'], width,
                    label='Research', color='lightgreen', alpha=0.7)
    bars3 = ax4.bar(x + width, scenario_scores['Industry'], width,
                    label='Industry', color='lightcoral', alpha=0.7)
    
    ax4.set_ylabel('Total Score', fontweight='bold')
    ax4.set_title('Model Selection by Application', fontweight='bold')
    ax4.set_xticks(x)
    ax4.set_xticklabels(model_names, rotation=45)
    ax4.legend()
    ax4.grid(True, alpha=0.3, axis='y')
    
    # Plot 5: Trade-off analysis (Accuracy vs Complexity)
    ax5 = axes[1, 1]
    accuracy_vals = [standard_scores[name]['accuracy'] for name in model_names]
    complexity_vals = [standard_scores[name]['complexity'] for name in model_names]
    
    scatter = ax5.scatter(complexity_vals, accuracy_vals, s=100, alpha=0.7, c=range(len(model_names)), cmap='viridis')
    
    # Add model labels
    for i, name in enumerate(model_names):
        ax5.annotate(name.split()[0], (complexity_vals[i], accuracy_vals[i]),
                    xytext=(5, 5), textcoords='offset points', fontsize=8)
    
    ax5.set_xlabel('Complexity Score', fontweight='bold')
    ax5.set_ylabel('Accuracy Score', fontweight='bold')
    ax5.set_title('Accuracy vs Complexity Trade-off', fontweight='bold')
    ax5.grid(True, alpha=0.3)
    
    # Add Pareto frontier concept
    ax5.text(0.02, 0.98, 'Pareto Optimal →', transform=ax5.transAxes,
            verticalalignment='top', fontweight='bold', color='red')
    
    # Plot 6: Recommendations summary
    ax6 = axes[1, 2]
    ax6.axis('off')
    
    # Find best models for each scenario
    best_standard = max(model_names, key=lambda x: standard_scores[x]['total_score'])
    best_research = max(model_names, key=lambda x: research_scores[x]['total_score'])
    best_industry = max(model_names, key=lambda x: industry_scores[x]['total_score'])
    
    recommendations_text = "Model Recommendations:\n\n"
    recommendations_text += f"🏆 Standard Use: {best_standard}\n"
    recommendations_text += f"   Score: {standard_scores[best_standard]['total_score']:.3f}\n\n"
    recommendations_text += f"🔬 Research: {best_research}\n"
    recommendations_text += f"   Score: {research_scores[best_research]['total_score']:.3f}\n\n"
    recommendations_text += f"🏭 Industry: {best_industry}\n"
    recommendations_text += f"   Score: {industry_scores[best_industry]['total_score']:.3f}\n\n"
    recommendations_text += f"\nKey Considerations:\n"
    recommendations_text += f"• Balance accuracy vs interpretability\n"
    recommendations_text += f"• Consider computational constraints\n"
    recommendations_text += f"• Evaluate robustness requirements"
    
    ax6.text(0.1, 0.9, recommendations_text, transform=ax6.transAxes,
             verticalalignment='top', fontfamily='monospace',
             bbox=dict(boxstyle='round', facecolor='lightgray', alpha=0.8))
    
    plt.tight_layout()
    plt.show()
    
    # Print detailed comparison
    print("\n📊 Detailed Model Comparison:")
    print("=" * 60)
    
    # Create comparison table
    header = "Model".ljust(20)
    for criteria in ['Accuracy', 'Robustness', 'Interpretability', 'Total']:
        header += f"{criteria.ljust(12)}"
    print(header)
    print("-" * len(header))
    
    for model_name in model_names:
        row = model_name.ljust(20)
        scores = standard_scores[model_name]
        row += f"{scores['accuracy']:.3f}  "
        row += f"{scores['robustness']:.3f}  "
        row += f"{scores['interpretability']:.3f}  "
        row += f"{scores['total_score']:.3f}  "
        print(row)
    
    print(f"\n🎯 Final Recommendations:")
    print(f"  • Standard applications: {best_standard}")
    print(f"  • Research-focused: {best_research}")
    print(f"  • Industry deployment: {best_industry}")
    
    return {
        'standard_scores': standard_scores,
        'research_scores': research_scores,
        'industry_scores': industry_scores,
        'recommendations': {
            'standard': best_standard,
            'research': best_research,
            'industry': best_industry
        }
    }

# Run comparative analysis demonstration
comparison_results = demonstrate_comparative_analysis()
```

## Chapter Summary

### Key Takeaways

1. **Comprehensive Metrics**: Multiple performance metrics (MSE, RMSE, MAE, R², NSE) provide different perspectives on model performance and should be used together for complete evaluation.

2. **Error Analysis**: Detailed error diagnostics help identify systematic biases, heteroscedasticity, and specific operating conditions where models perform poorly.

3. **Performance Maps**: Visual representations of motor performance across operating conditions enable intuitive understanding of model behavior and identification of optimal operating regions.

4. **Comparative Analysis**: Multi-criteria evaluation considering accuracy, robustness, efficiency, interpretability, and complexity helps select appropriate models for specific applications.

### Communication Strategies

- **Technical Audiences**: Use detailed metrics, error analysis, and statistical diagnostics
- **Management Audiences**: Focus on comparative rankings, trade-offs, and business impact
- **Engineering Teams**: Emphasize performance maps, operating regions, and practical implications

### Best Practices

- Always validate model performance on held-out test data
- Use multiple complementary metrics rather than relying on a single measure
- Perform error analysis to identify systematic issues and improvement opportunities
- Consider the specific requirements and constraints of your target application
- Document assumptions, limitations, and uncertainty estimates
